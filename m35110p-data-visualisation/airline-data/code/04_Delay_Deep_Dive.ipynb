## The Florida Aviation Divergence: Structural Delay Dynamics at MCO vs. MIA (2004–2008)

---

## 1. Executive Summary — Two Airports, Two Systems

Between 2004 and 2008, Florida’s two largest commercial aviation hubs—**Orlando International (MCO)** and **Miami International (MIA)**—followed sharply divergent operational trajectories.  
Despite serving the same state, overlapping carriers, and broadly similar macroeconomic conditions, the two airports exhibit **fundamentally different delay dynamics**.

The evidence shows that:

- **MCO behaves like a stable, capacity-managed system**, where delays are largely transient and scale modestly with traffic.
- **MIA behaves like a fragile, saturation-prone system**, where delays grow over time, respond nonlinearly to congestion, and propagate strongly across hours.

This divergence is not anecdotal. It is visible consistently across:
- time trends,
- congestion response curves,
- hourly backlog persistence,
- seasonal heatmaps,
- airline-level contributions.

The result is a widening **reliability gap** that cannot be explained by demand alone and instead reflects **structural differences in recovery capacity and operational resilience**.

---

## 2. Headline Metrics — What Changed, and Why It Matters

### 2.1 Long-Run Trend Divergence

We estimate monthly trends in mean arrival delay using HAC-robust inference.

- **MCO:**  
  - Trend ≈ **+0.03 min/month**  
  - Statistically indistinguishable from zero (\(p > 0.5\))  
  - Indicates *stationary performance*.

- **MIA:**  
  - Trend ≈ **+0.16 min/month**  
  - Highly significant (\(p < 0.001\))  
  - Implies a **compounding deterioration** in operational performance.

**Interpretation:**  
MIA’s delay growth is not cyclical noise—it reflects a system drifting away from equilibrium. MCO, by contrast, exhibits mean-reverting behavior.

---

### 2.2 Reliability Risk, Not Just Average Delay

While average delays differ, the **tail risk** differs far more.

- For flights exceeding ~150 minutes:
  - $P(\text{ArrDelay} \ge 15)$ rises sharply at **MIA**,
  - but remains comparatively flat at **MCO**.

This asymmetry indicates that **long-haul and inbound-dependent flights amplify existing congestion**, especially at MIA, where buffers are insufficient to absorb shocks.

---

## 3. Where Things Break: Statistical “Danger Zones”

### 3.1 Traffic Sensitivity and Congestion Regimes

Hourly traffic–delay analysis reveals two distinct regimes:

- **MCO:**  
  - Near-linear relationship between traffic and mean delay.
  - Delays increase, but predictably and proportionally.

- **MIA:**  
  - Strong **nonlinearity** once traffic reaches ~10–15 flights/hour.
  - Beyond this threshold, small traffic increases trigger large delay jumps.

Formally:
$
\text{Mean ArrDelay}_{MIA}
\approx \alpha + \gamma \cdot \text{Traffic} + \delta \cdot \mathbb{1}(\text{Peak})
$

After conditioning on hour and day:
- MIA incurs **≈ 4.6 additional minutes** of delay relative to MCO at comparable traffic levels.

**Interpretation:**  
This pattern is characteristic of a **binding ground-capacity constraint** (gates, taxiways, or ramp coordination), not airspace congestion alone.

---

### 3.2 Temporal Risk Concentration (Heatmap Evidence)

Heatmaps of arrival delay by hour, weekday, and month reveal systematic stress points:

- **Midnight Backlog (Hour 0):**  
  Elevated delays reflect unresolved congestion carried over from the prior day.

- **Hour-1 Void:**  
  A pronounced low-activity window (01:00) acts as a brief reset opportunity—but only if backlog is limited.

- **Peak Fragility:**  
  Thursday–Friday afternoons (14:00–19:00) are the most failure-prone periods, especially at MIA.

These patterns indicate that **delay accumulation, not random shocks**, drives poor performance.

---

### 3.3 The Summer Breakdown and Weather Amplification

Weather attribution analysis shows a sharp seasonal asymmetry:

- Weather-related delay share rises from ~2% to **~10%** in June–August.
- This increase is **not symmetric across airports**.

At MIA:
- July evenings (18:00–22:00) frequently exceed **30 minutes mean arrival delay**.
- Backlogs triggered by afternoon thunderstorms persist into late night.

At MCO:
- Weather shocks occur, but recovery is substantially faster.

**Interpretation:**  
Weather acts as a **trigger**, not the root cause. The real differentiator is **recovery capacity**—the ability to clear accumulated delays before the next operating cycle.

---

## 4. The Pareto of System Pain — Who Contributes to Delay?

Delay minutes are highly concentrated among a small number of carriers.

| Carrier | Total Late Minutes | Share of Total | Strategic Role |
|------|-------------------|---------------|----------------|
| **AA** | 3.91M | ~35% | Dominant hub carrier (MIA) |
| **WN** | 1.34M | ~12% | High-frequency operations |
| **DL** | 1.02M | ~9% | Network carrier |
| **FL / B6 / US** | — | — | Growth-oriented |

**Key insight:**  
American Airlines alone generates over **one-third of all delay minutes**. Given its dominance at MIA, this creates a **carrier–airport coupling** problem:  
inefficiencies at one reinforce congestion at the other.

This is not merely an airline issue—it is a **hub governance problem**.

---

## 5. Delay Persistence and Backlog Propagation (Core Structural Result)

### 5.1 Why Backlog Matters

Delays are not independent across time. Inbound aircraft, crews, and gates create **temporal dependencies**. To quantify this, we estimate:

$
\text{ArrDelay}_t
= \alpha + \beta \cdot \text{ArrDelay}_{t-1}
+ \text{controls} + \varepsilon_t
$

where $\beta$ measures **delay persistence**.

---

### 5.2 Visual Evidence: Propagation Curves

LOWESS curves of current-hour vs previous-hour delay show:

- Both airports lie below the $y=x$ line (full carryover),
- **MCO’s curve is steeper**—closer to full persistence.

This indicates that once delays occur at MCO, they tend to **carry forward more strongly** hour-to-hour.

---

### 5.3 Regression Estimates (Controlled)

| Airport | Backlog rate \(\beta\) | 95% CI |
|------|------------------|--------|
| **MCO** | **0.60** | [0.58, 0.62] |
| **MIA** | **0.46** | [0.44, 0.48] |

**Interpretation:**
- ~60% of MCO’s delay persists into the next hour.
- ~46% at MIA.

At first glance this seems counterintuitive—but it is crucial.

---

### 5.4 Reconciling Persistence vs Growth

The key distinction:

- **MCO:**  
  - High persistence **conditional on shocks**,  
  - but **fewer and smaller shocks overall**.

- **MIA:**  
  - Lower hour-to-hour persistence,  
  - but **larger shocks**, higher congestion sensitivity, and **worse recovery after peak periods**.

In short:
- **MCO** is a *stable system with memory*.
- **MIA** is an *unstable system with frequent resets—but at a higher baseline*.

---

## 6. Strategic Implications

### 6.1 Why Average Delay Is the Wrong Target

Two airports can share similar averages while exhibiting radically different risk profiles.  
What matters operationally is:
- tail risk,
- recovery speed,
- and backlog propagation.

Your analysis shows that **managing persistence and congestion regimes** yields higher returns than marginal traffic reduction.

---

## 7. Policy and Governance Interventions

### 7.1 Performance-Adjusted Landing Fees (PALF)

**Proposal:**  
Impose a 15–20% landing-fee surcharge on carriers in the bottom decile of hub-specific on-time performance.

- Encourages internalization of congestion externalities.
- Targets chronic contributors rather than all users.

---

### 7.2 Slot Recapture and Competitive Pressure

If a carrier:
- controls >50% of slots **and**
- contributes >30% of delay minutes,

then **forced slot divestiture** should be triggered.

Redistributing even 5% of peak slots can materially reduce systemic risk.

---

### 7.3 Time-of-Use Congestion Pricing

Heatmaps clearly identify **July evenings (18:00–21:00)** as a failure zone.

- 3× landing fees during these windows would:
  - flatten demand,
  - incentivize schedule shifting,
  - and reduce peak backlog formation.

---

### 7.4 Summer Resilience Buffers

Given weather-delay shares near 10% in summer:
- Mandate **20% longer turnarounds** for long-haul arrivals.
- Invest in high-speed taxi exits and A-CDM systems.

---

## 8. Final Takeaway

> **MCO and MIA are not just busier or quieter versions of the same system.  
They are structurally different machines.**

- MCO demonstrates controlled, predictable delay dynamics.
- MIA exhibits fragility, congestion amplification, and long-run deterioration.

The lesson is clear:
**Airport performance is governed less by volume than by structure, incentives, and recovery capacity.**


```python
import os
import math
import warnings

import numpy as np
import pandas as pd

import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots

from pandas.tseries.holiday import USFederalHolidayCalendar
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
from scipy import stats
from statsmodels.nonparametric.smoothers_lowess import lowess

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)

# -----------------------------
# Paths + constants
# -----------------------------
DATA_PATH = "../data/processed/mco_mia_clean.csv.gz"
OUTDIR = "../visualizations/2_Delay_Deep_Dive"
os.makedirs(OUTDIR, exist_ok=True)

AIRPORTS_LOCAL = "../data/airports.csv"
COUNTY_ZIP_LOCAL = "../data/cb_2018_us_county_500k.zip"

BLUE = "#1f77b4"  # MCO
RED = "#d62728"   # MIA

DOW_MAP = {1: "Mon", 2: "Tue", 3: "Wed", 4: "Thu", 5: "Fri", 6: "Sat", 7: "Sun"}
DOW_ORDER = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

MONTH_MAP = {
    1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr",
    5: "May", 6: "Jun", 7: "Jul", 8: "Aug",
    9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"
}
MONTH_ORDER = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

SEASON_ORDER = ["Winter", "Spring", "Summer", "Fall"]
SEASON_MAP = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Fall", 10: "Fall", 11: "Fall",
}

# -----------------------------
# Helpers
# -----------------------------
def nice_floor(x, step=5):
    return step * math.floor(x / step)

def nice_ceil(x, step=5):
    return step * math.ceil(x / step)

def robust_limits(values, qlo=0.01, qhi=0.99, pad_frac=0.08, step=5):
    values = np.asarray(values)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return -10, 10
    lo, hi = np.quantile(values, [qlo, qhi])
    pad = pad_frac * (hi - lo) if hi > lo else 5.0
    return nice_floor(lo - pad, step), nice_ceil(hi + pad, step)

def _mean_ci_normal(mean, sd, n, z=1.96):
    if n <= 1 or not np.isfinite(sd):
        return (np.nan, np.nan)
    se = sd / math.sqrt(n)
    return (mean - z * se, mean + z * se)

def save_plot(fig, stem: str):
    html_path = os.path.join(OUTDIR, f"{stem}.html")
    fig.write_html(html_path, include_plotlyjs="cdn")
    try:
        img_path = os.path.join(OUTDIR, f"{stem}.png")
        fig.write_image(img_path, scale=2)
    except Exception:
        pass
    return html_path

def save_text(text: str, filename: str):
    path = os.path.join(OUTDIR, filename)
    with open(path, "w", encoding="utf-8") as f:
        f.write(text)
    return path

def save_df(df: pd.DataFrame, filename: str):
    path = os.path.join(OUTDIR, filename)
    df.to_csv(path, index=False)
    return path

def thanksgiving_window(date_ts: pd.Timestamp) -> bool:
    year = int(date_ts.year)
    nov = pd.date_range(f"{year}-11-01", f"{year}-11-30", freq="D")
    tg = nov[nov.dayofweek == 3][3]  # 4th Thu
    return (date_ts >= tg - pd.Timedelta(days=1)) and (date_ts <= tg + pd.Timedelta(days=3))

# -----------------------------
# Load + core preprocessing
# -----------------------------
df = pd.read_csv(DATA_PATH, compression="gzip", low_memory=False)
df["FlightDate"] = pd.to_datetime(df.get("FlightDate", pd.NaT), errors="coerce")

for col in ["Cancelled", "Diverted"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)
    else:
        df[col] = 0

numeric_cols = [
    "ArrDelay", "Distance", "CRSElapsedTime", "ActualElapsedTime",
    "TaxiOut", "TaxiIn", "CarrierDelay", "WeatherDelay", "NASDelay",
    "SecurityDelay", "LateAircraftDelay",
    "Year", "Month", "DayOfWeek", "CRSDepTime_Hour",
]
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

if "Origin" not in df.columns:
    raise ValueError("Origin column not found.")
if "ArrDelay" not in df.columns:
    raise ValueError("ArrDelay column not found.")

d = df[df["Origin"].isin(["MCO", "MIA"])].copy()
d = d[(d["Cancelled"] == 0) & (d["Diverted"] == 0)].copy()
d = d[d["ArrDelay"].notna()].copy()

if "Year" in d.columns:
    d["Year"] = d["Year"].astype("Int64")
    d = d[(d["Year"] >= 2004) & (d["Year"] <= 2008)].copy()

d["ArrDelay_clip"] = d["ArrDelay"].clip(-60, 240)
d["IsArrDelayed15"] = (d["ArrDelay"] >= 15).astype(int)

d["DOW"] = d["DayOfWeek"].map(DOW_MAP) if "DayOfWeek" in d.columns else pd.NA

if "Month" in d.columns:
    d["MonthLabel"] = d["Month"].map(MONTH_MAP)
    d["Season"] = d["Month"].map(SEASON_MAP)
else:
    d["MonthLabel"] = pd.NA
    d["Season"] = pd.NA

d["CRSDepTime_Hour"] = pd.to_numeric(d.get("CRSDepTime_Hour", np.nan), errors="coerce")
d = d[d["CRSDepTime_Hour"].between(0, 23)].copy()
d["Hour"] = d["CRSDepTime_Hour"].astype(int)

d["YearMonth"] = d["FlightDate"].dt.to_period("M").dt.to_timestamp()

cal = USFederalHolidayCalendar()
holidays = cal.holidays(start="2004-01-01", end="2008-12-31")
d["USFedHoliday"] = d["FlightDate"].isin(holidays)

traffic = (
    d.groupby(["Origin", "FlightDate", "CRSDepTime_Hour"])
     .size()
     .reset_index(name="hourly_traffic")
)
d = d.merge(traffic, on=["Origin", "FlightDate", "CRSDepTime_Hour"], how="left")

# -----------------------------
# Event proxies
# -----------------------------
m = d["FlightDate"].dt.month
dom = d["FlightDate"].dt.day
dow0 = d["FlightDate"].dt.dayofweek  # Mon=0

d["Summer"] = m.isin([6, 7, 8])
d["XmasNewYear"] = ((m == 12) & (dom >= 20)) | ((m == 1) & (dom <= 5))
d["AfterSchool"] = (m.isin([9, 10])) & (dow0 <= 4) & (d["CRSDepTime_Hour"].between(15, 19))

unique_dates = pd.Series(pd.to_datetime(d["FlightDate"].dropna().unique()))
tg_map = {dt: thanksgiving_window(pd.Timestamp(dt)) for dt in unique_dates}
d["Thanksgiving"] = d["FlightDate"].map(tg_map).fillna(False)

wk = d["FlightDate"].dt.isocalendar().week.astype(int)
d["SpringBreak"] = ((m == 3) & (wk.between(10, 14))) | ((m == 4) & (wk.between(13, 14)))

event_order = ["Thanksgiving", "XmasNewYear", "USFedHoliday", "Summer", "SpringBreak", "AfterSchool"]

def assign_event(row):
    for e in event_order:
        if bool(row.get(e, False)):
            return e
    return "Normal"

d["Event"] = d.apply(assign_event, axis=1)

# -----------------------------
# Precompute tables
# -----------------------------
daily = (
    d.groupby(["Year", "Origin", "FlightDate"])["ArrDelay"]
     .mean()
     .reset_index(name="mean_arrdelay")
     .sort_values(["Year", "Origin", "FlightDate"])
)

daily_wide = (
    daily.groupby(["FlightDate", "Origin"])["mean_arrdelay"].mean().reset_index()
         .pivot(index="FlightDate", columns="Origin", values="mean_arrdelay")
         .dropna(subset=["MCO", "MIA"])
         .sort_index()
)

heat_dow_hour = (
    d.groupby(["DOW", "Hour"])["ArrDelay"].mean()
     .reset_index(name="mean_arrdelay")
     .pivot(index="DOW", columns="Hour", values="mean_arrdelay")
     .reindex(DOW_ORDER)
)

heat_month_hour = (
    d.groupby(["MonthLabel", "Hour"])["ArrDelay"].mean()
     .reset_index(name="mean_arrdelay")
     .pivot(index="MonthLabel", columns="Hour", values="mean_arrdelay")
     .reindex(MONTH_ORDER)
)

heat_season_airport = (
    d.groupby(["Season", "Origin"])["ArrDelay"].mean()
     .reset_index(name="mean_arrdelay")
     .pivot(index="Season", columns="Origin", values="mean_arrdelay")
     .reindex(SEASON_ORDER)
)

monthly = (
    d.groupby(["Origin", "YearMonth"])["ArrDelay"].mean()
     .reset_index(name="mean_arrdelay")
     .sort_values(["Origin", "YearMonth"])
)
monthly["t"] = (monthly["YearMonth"] - monthly["YearMonth"].min()).dt.days / 30.44

dur_delay = d.dropna(subset=["CRSElapsedTime", "ArrDelay"]).copy()
dur_delay["CRSElapsedTime"] = pd.to_numeric(dur_delay["CRSElapsedTime"], errors="coerce")
dur_delay = dur_delay.dropna(subset=["CRSElapsedTime"]).copy()
dur_delay = dur_delay[(dur_delay["CRSElapsedTime"] >= 20) & (dur_delay["CRSElapsedTime"] <= 400)].copy()
dur_delay["IsDelayed15"] = (dur_delay["ArrDelay"] >= 15).astype(int)

hourly = (
    d.dropna(subset=["FlightDate", "Hour", "ArrDelay"])
     .groupby(["Origin", "FlightDate", "Hour"])
     .agg(
        traffic=("ArrDelay", "size"),
        mean_delay=("ArrDelay", "mean"),
        p_delay=("ArrDelay", lambda x: float(np.mean(np.asarray(x) >= 15))),
     )
     .reset_index()
)

# -----------------------------
# Plot 1: daily diff (MCO - MIA) by year
# -----------------------------
years = sorted(daily["Year"].dropna().unique().tolist())
years = [int(y) for y in years]

diff_all = (daily_wide["MCO"] - daily_wide["MIA"]).dropna().values
ymin1, ymax1 = robust_limits(diff_all, qlo=0.01, qhi=0.99, pad_frac=0.10, step=2)

fig = make_subplots(
    rows=len(years), cols=1,
    shared_xaxes=True,
    vertical_spacing=0.03,
    subplot_titles=[str(y) for y in years],
)

for r, yr in enumerate(years, start=1):
    d_yr = daily[daily["Year"] == yr]

    mco = (
        d_yr[d_yr["Origin"] == "MCO"]
        .set_index("FlightDate")["mean_arrdelay"]
        .sort_index()
        .rename("MCO")
    )
    mia = (
        d_yr[d_yr["Origin"] == "MIA"]
        .set_index("FlightDate")["mean_arrdelay"]
        .sort_index()
        .rename("MIA")
    )

    aligned = pd.concat([mco, mia], axis=1).dropna().sort_index()
    if aligned.empty or len(aligned) < 2:
        continue

    diff = (aligned["MCO"] - aligned["MIA"]).astype(float)
    x = diff.index.map(lambda dt: pd.Timestamp(year=2000, month=dt.month, day=dt.day))
    y = diff.values

    y_mid = (y[:-1] + y[1:]) / 2.0
    seg_pos = y_mid >= 0

    x_pos, y_pos, x_neg, y_neg = [], [], [], []
    for i in range(len(y) - 1):
        if seg_pos[i]:
            x_pos += [x[i], x[i + 1], None]
            y_pos += [y[i], y[i + 1], None]
        else:
            x_neg += [x[i], x[i + 1], None]
            y_neg += [y[i], y[i + 1], None]

    show_leg = (r == 1)

    fig.add_trace(
        go.Scatter(
            x=x_pos, y=y_pos, mode="lines",
            line=dict(color=BLUE, width=2),
            name="diff ≥ 0",
            legendgroup="sign",
            showlegend=show_leg,
        ),
        row=r, col=1
    )
    fig.add_trace(
        go.Scatter(
            x=x_neg, y=y_neg, mode="lines",
            line=dict(color=RED, width=2),
            name="diff < 0",
            legendgroup="sign",
            showlegend=show_leg,
        ),
        row=r, col=1
    )

    fig.add_hline(y=0, line_dash="dash", line_width=1, line_color="black", opacity=0.35, row=r, col=1)
    fig.update_yaxes(range=[ymin1, ymax1], row=r, col=1)

fig.update_xaxes(tickformat="%b", dtick="M1")
fig.update_layout(
    title="Daily mean delay (minutes): MCO − MIA",
    template="simple_white",
    height=220 * max(1, len(years)),
    margin=dict(t=60, b=40),
)
fig.update_xaxes(title_text="Month", row=len(years), col=1)
save_plot(fig, "01_diff_by_year")

# -----------------------------
# Plot 2: heatmap DOW x hour
# -----------------------------
vals = heat_dow_hour.to_numpy().ravel()
vals = vals[np.isfinite(vals)]
vmin, vmax = (np.quantile(vals, 0.02), np.quantile(vals, 0.98)) if len(vals) else (0, 30)

fig = go.Figure(
    data=go.Heatmap(
        z=heat_dow_hour.values,
        x=list(heat_dow_hour.columns),
        y=list(heat_dow_hour.index),
        zmin=float(vmin),
        zmax=float(vmax),
        colorbar=dict(title="min"),
    )
)
fig.update_layout(
    title="Mean arrival delay",
    xaxis_title="Hour",
    yaxis_title="DOW",
    template="simple_white",
    height=420,
)
save_plot(fig, "02_heat_dow_hour")

s = heat_dow_hour.stack(dropna=True)
best_idx, best_val = s.idxmin(), float(s.min())
worst_idx, worst_val = s.idxmax(), float(s.max())
save_text(
    "\n".join([
        f"[DOW×Hour] best: {best_idx[0]} @ {int(best_idx[1]):02d}, mean={best_val:.2f} min",
        f"[DOW×Hour] worst: {worst_idx[0]} @ {int(worst_idx[1]):02d}, mean={worst_val:.2f} min",
    ]) + "\n",
    "02_heat_dow_hour_notes.txt"
)

# -----------------------------
# Plot 3: heatmap month x hour
# -----------------------------
vals = heat_month_hour.to_numpy().ravel()
vals = vals[np.isfinite(vals)]
vmin, vmax = (np.quantile(vals, 0.02), np.quantile(vals, 0.98)) if len(vals) else (0, 30)

fig = go.Figure(
    data=go.Heatmap(
        z=heat_month_hour.values,
        x=list(heat_month_hour.columns),
        y=list(heat_month_hour.index),
        zmin=float(vmin),
        zmax=float(vmax),
        colorbar=dict(title="min"),
    )
)
fig.update_layout(
    title="Mean arrival delay",
    xaxis_title="Hour",
    yaxis_title="Month",
    template="simple_white",
    height=420,
)
save_plot(fig, "03_heat_month_hour")

s = heat_month_hour.stack(dropna=True)
best_idx, best_val = s.idxmin(), float(s.min())
worst_idx, worst_val = s.idxmax(), float(s.max())
save_text(
    "\n".join([
        f"[Month×Hour] best: {best_idx[0]} @ {int(best_idx[1]):02d}, mean={best_val:.2f} min",
        f"[Month×Hour] worst: {worst_idx[0]} @ {int(worst_idx[1]):02d}, mean={worst_val:.2f} min",
    ]) + "\n",
    "03_heat_month_hour_notes.txt"
)

# -----------------------------
# Plot 4: MCO vs MIA daily mean
# -----------------------------
x = daily_wide["MCO"].values
y = daily_wide["MIA"].values
xmin, xmax = robust_limits(np.concatenate([x, y]), qlo=0.01, qhi=0.99, pad_frac=0.08, step=5)

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x, y=y,
    mode="markers",
    marker=dict(color="black", size=6, opacity=0.35),
    showlegend=False
))
fig.add_trace(go.Scatter(
    x=[xmin, xmax], y=[xmin, xmax],
    mode="lines",
    line=dict(color="gray", dash="dash", width=1),
    showlegend=False
))
fig.update_layout(
    title="Daily mean delay: MCO vs MIA",
    xaxis_title="MCO (min)",
    yaxis_title="MIA (min)",
    template="simple_white",
    width=650,
    height=650,
)
fig.update_xaxes(range=[xmin, xmax], scaleanchor="y", scaleratio=1)
fig.update_yaxes(range=[xmin, xmax])
save_plot(fig, "04_scatter_daily_mco_vs_mia")

save_text(
    f"[Scatter] n_days={len(daily_wide)}\nAbove diagonal: MIA worse. Below: MCO worse.\n",
    "04_scatter_notes.txt"
)

# -----------------------------
# Plot 5: monthly mean + trend + HAC tests
# -----------------------------
fig = go.Figure()
trend_rows = []
for ap, color in [("MCO", BLUE), ("MIA", RED)]:
    mth = monthly[monthly["Origin"] == ap].sort_values("YearMonth")

    fig.add_trace(go.Scatter(
        x=mth["YearMonth"], y=mth["mean_arrdelay"],
        mode="markers",
        marker=dict(color=color, size=7, opacity=0.45),
        name=ap,
    ))

    X_lr = mth[["t"]].values
    yv = mth["mean_arrdelay"].values
    model_lr = LinearRegression().fit(X_lr, yv)
    y_hat = model_lr.predict(X_lr)

    fig.add_trace(go.Scatter(
        x=mth["YearMonth"], y=y_hat,
        mode="lines",
        line=dict(color=color, width=3),
        showlegend=False
    ))

fig.update_layout(
    title="Monthly mean arrival delay",
    xaxis_title=None,
    yaxis_title="min",
    template="simple_white",
    height=450,
)
save_plot(fig, "05_monthly_mean_trend")

results = []
for ap in ["MCO", "MIA"]:
    mth = monthly[monthly["Origin"] == ap]
    X = sm.add_constant(mth["t"])
    yv = mth["mean_arrdelay"]
    fit = sm.OLS(yv, X).fit(cov_type="HAC", cov_kwds={"maxlags": 3})
    results.append({
        "Airport": ap,
        "Slope_min_per_month": float(fit.params["t"]),
        "StdErr": float(fit.bse["t"]),
        "t_stat": float(fit.tvalues["t"]),
        "p_value": float(fit.pvalues["t"]),
        "R2": float(fit.rsquared),
    })
trend_tests = pd.DataFrame(results)
save_df(trend_tests, "05_trend_tests_hac3.csv")

monthly_pooled = monthly.copy()
monthly_pooled["IsMIA"] = (monthly_pooled["Origin"] == "MIA").astype(int)
monthly_pooled["t_MIA"] = monthly_pooled["t"] * monthly_pooled["IsMIA"]

X = sm.add_constant(monthly_pooled[["t", "IsMIA", "t_MIA"]])
y = monthly_pooled["mean_arrdelay"]
model_diff = sm.OLS(y, X).fit(cov_type="HAC", cov_kwds={"maxlags": 3})
save_text(str(model_diff.summary()), "05_pooled_slope_diff_hac3.txt")

# -----------------------------
# Plot 6: duration vs P(delay>=15) + logit
# -----------------------------
dd = dur_delay.copy()
N_BINS = 20
dd["DurBin"] = pd.qcut(dd["CRSElapsedTime"], q=N_BINS, duplicates="drop")

g = (
    dd.groupby(["Origin", "DurBin"])
      .agg(
          p_delay=("IsDelayed15", "mean"),
          mean_dur=("CRSElapsedTime", "mean"),
          n=("IsDelayed15", "size"),
      )
      .reset_index()
)

fig = go.Figure()
for ap, color in [("MCO", BLUE), ("MIA", RED)]:
    sub = g[g["Origin"] == ap].sort_values("mean_dur")
    fig.add_trace(go.Scatter(
        x=sub["mean_dur"], y=sub["p_delay"],
        mode="lines+markers",
        line=dict(color=color, width=3),
        marker=dict(size=7),
        name=ap
    ))
fig.update_layout(
    title="P(arrival delay ≥ 15) vs duration",
    xaxis_title="Duration (min)",
    yaxis_title="Probability",
    template="simple_white",
    height=450,
)
save_plot(fig, "06_duration_prob_delay")

X = pd.DataFrame({
    "Dur_30min": dd["CRSElapsedTime"] / 30.0,
    "IsMIA": (dd["Origin"] == "MIA").astype(int),
})
X = sm.add_constant(X, has_constant="add")
y = dd["IsDelayed15"]
clusters = dd["FlightDate"]

logit = sm.Logit(y, X).fit(cov_type="cluster", cov_kwds={"groups": clusters}, disp=False)

params = logit.params
conf = logit.conf_int()
odds = pd.DataFrame({
    "term": params.index.astype(str),
    "OR": np.exp(params.values),
    "CI_low": np.exp(conf[0].values),
    "CI_high": np.exp(conf[1].values),
})
save_df(odds, "06_logit_odds_clustered.csv")
save_text(str(logit.summary()), "06_logit_summary.txt")

# -----------------------------
# Plot 7: traffic vs mean delay + correlations + OLS
# -----------------------------
h = hourly.copy()
N_BINS = 20
h["TrafficBin"] = h.groupby("Origin")["traffic"].transform(lambda x: pd.qcut(x, q=N_BINS, duplicates="drop"))

gb = (
    h.groupby(["Origin", "TrafficBin"])
     .agg(
         mean_traffic=("traffic", "mean"),
         mean_delay=("mean_delay", "mean"),
         p_delay=("p_delay", "mean"),
         n=("traffic", "size"),
     )
     .reset_index()
)

fig = go.Figure()
for ap, color in [("MCO", BLUE), ("MIA", RED)]:
    sub = gb[gb["Origin"] == ap].sort_values("mean_traffic")
    fig.add_trace(go.Scatter(
        x=sub["mean_traffic"], y=sub["mean_delay"],
        mode="lines+markers",
        line=dict(color=color, width=3),
        marker=dict(size=7),
        name=ap
    ))
fig.update_layout(
    title="Mean arrival delay vs traffic",
    xaxis_title="Traffic (flights/hour)",
    yaxis_title="min",
    template="simple_white",
    height=450,
)
save_plot(fig, "07_traffic_mean_delay")

rows = []
for ap in ["MCO", "MIA"]:
    sub = h[h["Origin"] == ap]
    r_mean, p_mean = stats.spearmanr(sub["traffic"], sub["mean_delay"])
    r_prob, p_prob = stats.spearmanr(sub["traffic"], sub["p_delay"])
    rows.append({
        "Airport": ap,
        "Spearman_r_mean": float(r_mean),
        "p_value_mean": float(p_mean),
        "Spearman_r_prob": float(r_prob),
        "p_value_prob": float(p_prob),
    })
corr_tbl = pd.DataFrame(rows)
save_df(corr_tbl, "07_spearman_hourly.csv")

m_ols = h.copy()
m_ols["DOWnum"] = m_ols["FlightDate"].dt.dayofweek
X = pd.get_dummies(m_ols[["traffic", "Hour", "DOWnum", "Origin"]], drop_first=True)
X = sm.add_constant(X, has_constant="add")
y = m_ols["mean_delay"]
ols = sm.OLS(y, X).fit(cov_type="cluster", cov_kwds={"groups": m_ols["FlightDate"]})
save_text(str(ols.summary()), "07_ols_mean_delay_clustered.txt")

# -----------------------------
# Plot 7B: backlog / propagation (lagged mean_delay)
# -----------------------------
req = {"Origin", "FlightDate", "Hour", "traffic", "mean_delay"}
miss = req - set(hourly.columns)
if miss:
    raise ValueError(f"hourly missing columns: {sorted(miss)}")

h = hourly.copy()
h = h[h["Origin"].isin(["MCO", "MIA"])].copy()

h["FlightDate"] = pd.to_datetime(h["FlightDate"], errors="coerce")
h["Hour"] = pd.to_numeric(h["Hour"], errors="coerce")
h["traffic"] = pd.to_numeric(h["traffic"], errors="coerce")
h["mean_delay"] = pd.to_numeric(h["mean_delay"], errors="coerce")
h = h.dropna(subset=["Origin", "FlightDate", "Hour", "traffic", "mean_delay"]).copy()
h["Hour"] = h["Hour"].astype(int)

h = h.sort_values(["Origin", "FlightDate", "Hour"]).copy()
h["lag_mean_delay"] = h.groupby(["Origin", "FlightDate"])["mean_delay"].shift(1)
hl = h.dropna(subset=["lag_mean_delay"]).copy()

def _trim_quantiles(df, col, qlo=0.01, qhi=0.99):
    lo, hi = df[col].quantile([qlo, qhi])
    return df[(df[col] >= lo) & (df[col] <= hi)].copy()

hl_parts = []
for ap in ["MCO", "MIA"]:
    sub = hl[hl["Origin"] == ap].copy()
    if len(sub) < 100:
        continue
    sub = _trim_quantiles(sub, "lag_mean_delay", 0.01, 0.99)
    sub = _trim_quantiles(sub, "mean_delay", 0.01, 0.99)
    hl_parts.append(sub)

if not hl_parts:
    raise ValueError("Not enough lagged observations after trimming.")
hl = pd.concat(hl_parts, ignore_index=True)

fig = go.Figure()
for ap, color in [("MCO", BLUE), ("MIA", RED)]:
    sub = hl[hl["Origin"] == ap]
    xlag = sub["lag_mean_delay"].to_numpy(float)
    ycur = sub["mean_delay"].to_numpy(float)

    smth = lowess(ycur, xlag, frac=0.25, it=0, return_sorted=True)
    xs, ys = smth[:, 0], smth[:, 1]

    fig.add_trace(go.Scatter(
        x=xlag, y=ycur,
        mode="markers",
        marker=dict(color=color, size=5, opacity=0.10),
        showlegend=False
    ))
    fig.add_trace(go.Scatter(
        x=xs, y=ys,
        mode="lines",
        line=dict(color=color, width=4),
        name=ap
    ))

allv = np.concatenate([hl["lag_mean_delay"].to_numpy(float), hl["mean_delay"].to_numpy(float)])
xmin = float(np.quantile(allv, 0.01))
xmax = float(np.quantile(allv, 0.99))

fig.add_trace(go.Scatter(
    x=[xmin, xmax], y=[xmin, xmax],
    mode="lines",
    line=dict(color="black", width=1, dash="dash"),
    showlegend=False
))

fig.update_layout(
    title="Backlog: mean_delay(t) vs mean_delay(t−1)",
    xaxis_title="t−1 (min)",
    yaxis_title="t (min)",
    template="simple_white",
    height=520,
)
fig.update_xaxes(range=[xmin, xmax])
fig.update_yaxes(range=[xmin, xmax], scaleanchor="x", scaleratio=1)
save_plot(fig, "07B_backlog_lowess")

mreg = hl.copy()
mreg["DOWnum"] = mreg["FlightDate"].dt.dayofweek
mreg["IsMIA"] = (mreg["Origin"] == "MIA").astype(int)
mreg["lag_x_IsMIA"] = mreg["lag_mean_delay"] * mreg["IsMIA"]

X = mreg[["lag_mean_delay", "lag_x_IsMIA", "traffic", "Hour", "DOWnum", "IsMIA"]].copy()
X = sm.add_constant(X, has_constant="add")
y = mreg["mean_delay"]

fit = sm.OLS(y, X).fit(cov_type="cluster", cov_kwds={"groups": mreg["FlightDate"]})
save_text(str(fit.summary()), "07B_backlog_regression_clustered.txt")

beta_mco = float(fit.params["lag_mean_delay"])
beta_mia = float(fit.params["lag_mean_delay"] + fit.params["lag_x_IsMIA"])

cov = fit.cov_params()
var_mco = float(cov.loc["lag_mean_delay", "lag_mean_delay"])
var_mia = float(
    cov.loc["lag_mean_delay", "lag_mean_delay"]
    + cov.loc["lag_x_IsMIA", "lag_x_IsMIA"]
    + 2.0 * cov.loc["lag_mean_delay", "lag_x_IsMIA"]
)

se_mco = float(np.sqrt(max(var_mco, 0.0)))
se_mia = float(np.sqrt(max(var_mia, 0.0)))

z = 1.96
ci_mco = (beta_mco - z * se_mco, beta_mco + z * se_mco)
ci_mia = (beta_mia - z * se_mia, beta_mia + z * se_mia)

beta_tbl = pd.DataFrame([
    {"Airport": "MCO", "beta": beta_mco, "ci_low": ci_mco[0], "ci_high": ci_mco[1]},
    {"Airport": "MIA", "beta": beta_mia, "ci_low": ci_mia[0], "ci_high": ci_mia[1]},
])
save_df(beta_tbl, "07B_backlog_beta_ci.csv")

figb = go.Figure()
for ap, color in [("MCO", BLUE), ("MIA", RED)]:
    row = beta_tbl[beta_tbl["Airport"] == ap].iloc[0]
    figb.add_trace(go.Scatter(
        x=[ap],
        y=[row["beta"]],
        mode="markers",
        marker=dict(color=color, size=14),
        error_y=dict(
            type="data",
            symmetric=False,
            array=[row["ci_high"] - row["beta"]],
            arrayminus=[row["beta"] - row["ci_low"]],
            thickness=2,
            width=6,
        ),
        showlegend=False
    ))
figb.add_hline(y=0, line_dash="dash", line_color="black", opacity=0.35)
figb.update_layout(
    title="Backlog rate (β)",
    xaxis_title=None,
    yaxis_title="β",
    template="simple_white",
    height=420,
)
save_plot(figb, "07B_backlog_beta")

# -----------------------------
# Plot 8: airline mean + 95% CI
# -----------------------------
airline_col_candidates = ["Reporting_Airline", "Carrier", "UniqueCarrier", "OP_UNIQUE_CARRIER"]
airline_col = next((c for c in airline_col_candidates if c in d.columns), None)

if airline_col is not None:
    air = d.dropna(subset=[airline_col, "ArrDelay"]).copy()
    MIN_N = 500
    TOP_K = 15

    stats_air = (
        air.groupby(airline_col)["ArrDelay"]
           .agg(mean="mean", sd="std", n="size")
           .reset_index()
    )
    stats_air = stats_air[stats_air["n"] >= MIN_N].copy()

    if not stats_air.empty:
        stats_air["ci_low"], stats_air["ci_high"] = zip(*stats_air.apply(
            lambda r: _mean_ci_normal(float(r["mean"]), float(r["sd"]), int(r["n"])), axis=1
        ))
        stats_air = stats_air.sort_values("mean", ascending=False).head(TOP_K).copy()

        fig = go.Figure()
        fig.add_trace(go.Scatter(
            x=stats_air["mean"],
            y=stats_air[airline_col].astype(str),
            mode="markers",
            marker=dict(color="black", size=8),
            error_x=dict(
                type="data",
                symmetric=False,
                array=stats_air["ci_high"] - stats_air["mean"],
                arrayminus=stats_air["mean"] - stats_air["ci_low"],
                thickness=2,
                width=4,
            ),
            showlegend=False
        ))
        fig.add_vline(x=0, line_dash="dash", line_width=1, line_color="black", opacity=0.4)
        fig.update_layout(
            title="Mean arrival delay by airline",
            xaxis_title="min",
            yaxis_title=None,
            template="simple_white",
            height=80 * len(stats_air) + 180,
        )
        save_plot(fig, "08_airline_mean_ci")

        out_tbl = stats_air[[airline_col, "mean", "ci_low", "ci_high", "n"]].copy()
        out_tbl = out_tbl.rename(columns={"mean": "mean_arrdelay", "n": "n_flights"})
        save_df(out_tbl, "08_airline_ci_table.csv")

# -----------------------------
# Plot 9: Pareto total late minutes by airline
# -----------------------------
if airline_col is not None:
    air_p = d.dropna(subset=[airline_col, "ArrDelay"]).copy()
    air_p["ArrDelay_pos"] = air_p["ArrDelay"].clip(lower=0)

    MIN_N = 500
    g_pareto = (
        air_p.groupby(airline_col)
             .agg(
                 total_delay_min=("ArrDelay_pos", "sum"),
                 n_flights=("ArrDelay_pos", "size"),
             )
             .reset_index()
    )
    g_pareto = g_pareto[g_pareto["n_flights"] >= MIN_N].copy()
    g_pareto = g_pareto.sort_values("total_delay_min", ascending=False).reset_index(drop=True)

    total_delay = float(g_pareto["total_delay_min"].sum())
    if total_delay > 0 and len(g_pareto):
        g_pareto["cum_delay_min"] = g_pareto["total_delay_min"].cumsum()
        g_pareto["cum_pct"] = 100.0 * g_pareto["cum_delay_min"] / total_delay
        save_df(g_pareto, "09_pareto_table.csv")

        fig = go.Figure()
        fig.add_trace(go.Bar(
            x=g_pareto[airline_col].astype(str),
            y=g_pareto["total_delay_min"],
            marker_color="black",
            name="Total late minutes",
            yaxis="y1",
        ))
        fig.add_trace(go.Scatter(
            x=g_pareto[airline_col].astype(str),
            y=g_pareto["cum_pct"],
            mode="lines+markers",
            line=dict(color="black", width=3),
            marker=dict(size=7),
            name="Cum %",
            yaxis="y2",
        ))
        fig.add_hline(y=80, line_dash="dash", line_color="black", opacity=0.5, yref="y2")
        fig.update_layout(
            title="Pareto: late-arrival minutes by airline",
            template="simple_white",
            xaxis=dict(title=None, tickangle=45),
            yaxis=dict(title="minutes", showgrid=False),
            yaxis2=dict(
                title="%",
                overlaying="y",
                side="right",
                range=[0, 105],
                showgrid=False,
            ),
            height=520,
        )
        save_plot(fig, "09_pareto_airline")

        n80 = int((g_pareto["cum_pct"] <= 80).sum()) + 1
        save_text(
            f"[Pareto] airlines to reach ~80%: {n80} / {len(g_pareto)}\n",
            "09_pareto_notes.txt"
        )

# -----------------------------
# Plot 10: weather share by month
# -----------------------------
w = d.copy()
for c in ["CarrierDelay", "WeatherDelay", "NASDelay", "SecurityDelay", "LateAircraftDelay"]:
    if c not in w.columns:
        w[c] = np.nan

w["WeatherDelay"] = w["WeatherDelay"].fillna(0)
w["TotalDelayCauseMin"] = (
    w["CarrierDelay"].fillna(0) +
    w["WeatherDelay"] +
    w["NASDelay"].fillna(0) +
    w["SecurityDelay"].fillna(0) +
    w["LateAircraftDelay"].fillna(0)
)
w = w[w["TotalDelayCauseMin"] > 0].copy()
w["WeatherShare"] = w["WeatherDelay"] / w["TotalDelayCauseMin"]

g_weather = (
    w.groupby("MonthLabel")["WeatherShare"].mean()
     .reindex(MONTH_ORDER)
)
weather_tbl = pd.DataFrame({"Month": g_weather.index, "WeatherShare_pct": 100 * g_weather.values})
save_df(weather_tbl, "10_weather_share_by_month.csv")

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=g_weather.index,
    y=100 * g_weather.values,
    mode="lines+markers",
    line=dict(color="black", width=3),
    marker=dict(size=7),
    showlegend=False
))
fig.update_layout(
    title="Weather share of delay minutes",
    xaxis_title=None,
    yaxis_title="%",
    template="simple_white",
    height=420,
)
save_plot(fig, "10_weather_share_by_month")

# -----------------------------
# Plot 11: LOWESS traffic vs delay by event
# -----------------------------
event_colors = {
    "Normal": "#000000",
    "Summer": "#ff7f0e",
    "XmasNewYear": "#d62728",
    "Thanksgiving": "#8c564b",
    "USFedHoliday": "#9467bd",
    "SpringBreak": "#2ca02c",
    "AfterSchool": "#1f77b4",
}

fig = go.Figure()
plotted_any = False

for ev, color in event_colors.items():
    pe = d[(d["Event"] == ev)].dropna(subset=["hourly_traffic", "ArrDelay_clip"]).copy()
    if pe.shape[0] < 800:
        continue

    smth = lowess(
        pe["ArrDelay_clip"].values,
        pe["hourly_traffic"].values,
        frac=0.25,
        it=0,
        return_sorted=True
    )
    fig.add_trace(go.Scatter(
        x=smth[:, 0], y=smth[:, 1],
        mode="lines",
        line=dict(color=color, width=4),
        name=ev
    ))
    plotted_any = True

fig.update_layout(
    title="Traffic → delay (LOWESS)",
    xaxis_title="Traffic",
    yaxis_title="Delay (min)",
    template="simple_white",
    height=560,
)
if plotted_any:
    save_plot(fig, "11_lowess_by_event")
else:
    save_text("[LOWESS by event] no groups met the sample threshold.\n", "11_lowess_by_event_notes.txt")

# -----------------------------
# Plot 12: heatmap season x airport
# -----------------------------
vals = heat_season_airport.to_numpy().ravel()
vals = vals[np.isfinite(vals)]
vmin, vmax = (np.quantile(vals, 0.02), np.quantile(vals, 0.98)) if len(vals) else (0, 30)

fig = go.Figure(
    data=go.Heatmap(
        z=heat_season_airport.values,
        x=list(heat_season_airport.columns),
        y=list(heat_season_airport.index),
        zmin=float(vmin),
        zmax=float(vmax),
        colorbar=dict(title="min"),
    )
)
fig.update_layout(
    title="Mean arrival delay",
    xaxis_title=None,
    yaxis_title=None,
    template="simple_white",
    height=420,
)
save_plot(fig, "12_heat_season_airport")

# -----------------------------
# Save key tables for reuse
# -----------------------------
save_df(daily, "tables_daily_mean_arrdelay.csv")
save_df(daily_wide.reset_index(), "tables_daily_wide_mco_mia.csv")
save_df(monthly, "tables_monthly_mean_arrdelay.csv")
save_df(hourly, "tables_hourly_aggregates.csv")
save_df(heat_dow_hour.reset_index().melt(id_vars="DOW", var_name="Hour", value_name="mean_arrdelay"), "tables_heat_dow_hour_long.csv")
save_df(heat_month_hour.reset_index().melt(id_vars="MonthLabel", var_name="Hour", value_name="mean_arrdelay"), "tables_heat_month_hour_long.csv")
save_df(heat_season_airport.reset_index().melt(id_vars="Season", var_name="Origin", value_name="mean_arrdelay"), "tables_heat_season_airport_long.csv")

save_text("DONE (ArrDelay pipeline). Outputs saved to OUTDIR.\n", "run_log.txt")
```

```python
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import geopandas as gpd
except ImportError as e:
    raise ImportError("geopandas is required for this cell (pip install geopandas).") from e

from shapely.geometry import Point
from matplotlib.lines import Line2D
from matplotlib.patches import Arc
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

# -----------------------------
# Config
# -----------------------------
OUTDIR = globals().get("OUTDIR", "../visualizations/2_Delay_Deep_Dive")
os.makedirs(OUTDIR, exist_ok=True)

AIRPORTS_LOCAL = globals().get("AIRPORTS_LOCAL", "../data/airports.csv")
COUNTY_ZIP_LOCAL = globals().get("COUNTY_ZIP_LOCAL", "../data/cb_2018_us_county_500k.zip")

AIRLINE_COL_CANDIDATES = ["UniqueCarrier", "OP_UNIQUE_CARRIER", "Carrier", "AIRLINE"]
MIN_COUNTY_FLIGHTS = 500

HALF_RING_KM = [1000, 2000, 3000]
SIZE_MULTIPLIER = 2.6

# -----------------------------
# Helpers
# -----------------------------
def _zfill_fips(x):
    return (
        pd.Series(x).astype(str)
        .str.replace(r"\.0$", "", regex=True)
        .str.replace(r"\s+", "", regex=True)
        .replace({"nan": np.nan, "None": np.nan, "": np.nan})
        .str.zfill(5)
    )

def _standardize_airport_code(series):
    return (
        series.astype(str)
        .str.strip()
        .str.upper()
        .replace({"NAN": np.nan, "NONE": np.nan, "": np.nan})
    )

def _safe_sjoin(left, right, how="left", predicate="within"):
    try:
        return gpd.sjoin(left, right, how=how, predicate=predicate)
    except TypeError:
        return gpd.sjoin(left, right, how=how, op=predicate)

def _pick_airline_col(df: pd.DataFrame) -> str:
    for c in AIRLINE_COL_CANDIDATES:
        if c in df.columns:
            return c
    raise ValueError(f"No airline column found. Tried: {AIRLINE_COL_CANDIDATES}")

def _conus_filter(counties_gdf: gpd.GeoDataFrame) -> gpd.GeoDataFrame:
    if "STATEFP" not in counties_gdf.columns:
        raise ValueError("County file must include STATEFP to drop AK/HI/territories.")
    exclude = {"02", "15", "60", "66", "69", "72", "78"}  # AK, HI, AS, GU, MP, PR, VI
    return counties_gdf[~counties_gdf["STATEFP"].astype(str).isin(exclude)].copy()

def save_fig(fig, stem: str, dpi: int = 200):
    png_path = os.path.join(OUTDIR, f"{stem}.png")
    pdf_path = os.path.join(OUTDIR, f"{stem}.pdf")
    fig.savefig(png_path, dpi=dpi, bbox_inches="tight")
    fig.savefig(pdf_path, bbox_inches="tight")
    return png_path, pdf_path

def save_df(df: pd.DataFrame, filename: str):
    path = os.path.join(OUTDIR, filename)
    df.to_csv(path, index=False)
    return path

def save_text(text: str, filename: str):
    path = os.path.join(OUTDIR, filename)
    with open(path, "w", encoding="utf-8") as f:
        f.write(text)
    return path

# -----------------------------
# Inputs
# -----------------------------
if "d" not in globals():
    raise ValueError("Expected DataFrame `d` in memory.")
d_in = d

airport_key = None
if "Dest" in d_in.columns:
    airport_key = "Dest"
elif "DEST" in d_in.columns:
    airport_key = "DEST"
elif "Origin" in d_in.columns:
    airport_key = "Origin"
else:
    raise ValueError("Neither Dest nor Origin found in `d`.")

dest_candidates = ["DestCountyFips", "DEST_COUNTY_FIPS", "DestFIPS", "DEST_FIPS", "DEST_COUNTY_FIPS_CODE"]
orig_candidates = ["OriginCountyFips", "ORIGIN_COUNTY_FIPS", "OriginFIPS", "ORIGIN_FIPS"]
generic_candidates = ["CountyFips", "COUNTY_FIPS", "FIPS", "fips"]

fips_col = next((c for c in dest_candidates if c in d_in.columns), None)
fips_role = "Destination"
if fips_col is None:
    if airport_key.lower().startswith("orig"):
        fips_col = next((c for c in orig_candidates if c in d_in.columns), None)
        fips_role = "Origin"
    if fips_col is None:
        fips_col = next((c for c in generic_candidates if c in d_in.columns), None)
        fips_role = "County"

d_map = d_in.copy()

# -----------------------------
# County GEOID mapping
# -----------------------------
if fips_col is None:
    ap = pd.read_csv(AIRPORTS_LOCAL, low_memory=False)
    required_cols = {"iata_code", "latitude_deg", "longitude_deg"}
    missing = required_cols - set(ap.columns)
    if missing:
        raise ValueError(f"airports.csv missing columns: {sorted(missing)}")

    ap = ap[["iata_code", "latitude_deg", "longitude_deg", "name"]].copy()
    ap = ap.rename(columns={"iata_code": "Airport", "latitude_deg": "Latitude", "longitude_deg": "Longitude"})
    ap["Airport"] = _standardize_airport_code(ap["Airport"])
    ap = ap.dropna(subset=["Airport", "Latitude", "Longitude"]).copy()

    counties_raw = gpd.read_file(f"zip://{COUNTY_ZIP_LOCAL}")
    if counties_raw.crs is None:
        counties_raw = counties_raw.set_crs("EPSG:4269", allow_override=True)
    counties_raw["GEOID"] = counties_raw["GEOID"].astype(str)
    counties_raw = _conus_filter(counties_raw)
    counties_4326 = counties_raw.to_crs("EPSG:4326")[["GEOID", "geometry"]].copy()

    ap_gdf = gpd.GeoDataFrame(
        ap,
        geometry=gpd.points_from_xy(ap["Longitude"], ap["Latitude"]),
        crs="EPSG:4326",
    )
    ap_join = _safe_sjoin(ap_gdf, counties_4326, how="left", predicate="within")
    ap_to_geoid = ap_join[["Airport", "GEOID"]].dropna(subset=["GEOID"]).drop_duplicates("Airport")

    d_map[airport_key] = _standardize_airport_code(d_map[airport_key])
    d_map = d_map.merge(
        ap_to_geoid.rename(columns={"Airport": airport_key, "GEOID": "DerivedCountyGEOID"}),
        on=airport_key,
        how="left",
    )

    fips_col = "DerivedCountyGEOID"
    fips_role = f"Derived({airport_key})"

for req in ["ArrDelay", "FlightDate"]:
    if req not in d_map.columns:
        raise ValueError(f"{req} not found in `d`.")

airline_col = _pick_airline_col(d_map)

dd = d_map.dropna(subset=[fips_col, "ArrDelay", "FlightDate", airline_col]).copy()
dd["GEOID"] = _zfill_fips(dd[fips_col])
dd[airline_col] = (
    dd[airline_col].astype(str).str.strip().str.upper()
    .replace({"NAN": np.nan, "NONE": np.nan, "": np.nan})
)
dd = dd.dropna(subset=["GEOID", airline_col])

# Day-level mean -> flight-weighted across days
county_day = (
    dd.groupby(["GEOID", "FlightDate"])
    .agg(mean_delay=("ArrDelay", "mean"), n_flights=("ArrDelay", "size"))
    .reset_index()
)

county_delay = (
    county_day.groupby("GEOID")
    .apply(lambda x: np.average(x["mean_delay"], weights=x["n_flights"]))
    .reset_index(name="county_mean_delay")
)

county_n = (
    county_day.groupby("GEOID")["n_flights"]
    .sum()
    .reset_index(name="county_total_flights")
)

county_delay = county_delay.merge(county_n, on="GEOID", how="left")
county_delay.loc[county_delay["county_total_flights"] < MIN_COUNTY_FLIGHTS, "county_mean_delay"] = np.nan

if county_delay["county_mean_delay"].notna().sum() == 0:
    raise ValueError("No counties left after filtering. Reduce MIN_COUNTY_FLIGHTS or check mapping.")

county_carrier = (
    dd.groupby(["GEOID", airline_col])
    .size()
    .reset_index(name="carrier_flights")
)

county_dominant = (
    county_carrier.sort_values(["GEOID", "carrier_flights"], ascending=[True, False])
    .drop_duplicates(subset=["GEOID"], keep="first")
    .rename(columns={airline_col: "dominant_carrier"})
    .copy()
)

save_df(county_delay, "map_county_delay.csv")
save_df(county_dominant, "map_county_dominant_carrier.csv")

# -----------------------------
# Counties geometry + merge metrics
# -----------------------------
counties = gpd.read_file(f"zip://{COUNTY_ZIP_LOCAL}")
if counties.crs is None:
    counties = counties.set_crs("EPSG:4269", allow_override=True)
counties = _conus_filter(counties)

counties_5070 = counties.to_crs("EPSG:5070").copy()
counties_5070["GEOID"] = counties_5070["GEOID"].astype(str)

gdf = counties_5070.merge(county_delay, on="GEOID", how="left")

# -----------------------------
# Airports + midpoint (EPSG:5070)
# -----------------------------
fallback_coords = {"MCO": (28.4312, -81.3081), "MIA": (25.7959, -80.2870)}

ap = pd.read_csv(AIRPORTS_LOCAL, low_memory=False)
ap = ap.rename(columns={"iata_code": "Airport", "latitude_deg": "Latitude", "longitude_deg": "Longitude"})
ap["Airport"] = _standardize_airport_code(ap["Airport"])
ap = ap.dropna(subset=["Airport", "Latitude", "Longitude"]).copy()

rows = []
for code in ["MCO", "MIA"]:
    hit = ap.loc[ap["Airport"] == code, ["Latitude", "Longitude"]].head(1)
    if len(hit) == 1:
        lat = float(hit["Latitude"].iloc[0])
        lon = float(hit["Longitude"].iloc[0])
    else:
        lat, lon = fallback_coords[code]
    rows.append({"Airport": code, "Latitude": lat, "Longitude": lon})

airports_pts = pd.DataFrame(rows)
airports_gdf = gpd.GeoDataFrame(
    airports_pts,
    geometry=gpd.points_from_xy(airports_pts["Longitude"], airports_pts["Latitude"]),
    crs="EPSG:4326",
).to_crs("EPSG:5070")

p_mco = airports_gdf.loc[airports_gdf["Airport"] == "MCO", "geometry"].iloc[0]
p_mia = airports_gdf.loc[airports_gdf["Airport"] == "MIA", "geometry"].iloc[0]
midpoint = Point((p_mco.x + p_mia.x) / 2.0, (p_mco.y + p_mia.y) / 2.0)

# -----------------------------
# Dominant-carrier marker points
# -----------------------------
rep = gdf[["GEOID", "geometry"]].copy()
rep["geometry"] = rep["geometry"].representative_point()

carrier_pts = county_dominant.merge(rep, on="GEOID", how="inner")
carrier_pts_gdf = gpd.GeoDataFrame(carrier_pts, geometry="geometry", crs="EPSG:5070")
if carrier_pts_gdf.empty:
    raise ValueError("No carrier points to plot. Check airline column and county mapping.")

top_carriers_overall = (
    county_carrier.groupby(airline_col)["carrier_flights"].sum()
    .sort_values(ascending=False)
    .head(8)
    .index.tolist()
)

marker_cycle = ["o", "s", "^", "D", "P", "X", "v", "*"]
carrier_to_marker = {c: marker_cycle[i] for i, c in enumerate(top_carriers_overall)}

carrier_pts_gdf["marker"] = carrier_pts_gdf["dominant_carrier"].map(carrier_to_marker).fillna(".")
cf = carrier_pts_gdf["carrier_flights"].fillna(0).to_numpy()
cf_sqrt = np.sqrt(np.maximum(cf, 0))
smin, smax = 40, 300
cf_norm = (cf_sqrt - cf_sqrt.min()) / (cf_sqrt.max() - cf_sqrt.min() + 1e-12)
carrier_pts_gdf["size"] = (smin + cf_norm * (smax - smin)) * SIZE_MULTIPLIER

save_text(
    "\n".join([f"{k},{v}" for k, v in carrier_to_marker.items()]) + "\n",
    "map_carrier_marker_mapping.csv",
)

# -----------------------------
# Plot
# -----------------------------
vals = gdf["county_mean_delay"].dropna().values
vmin, vmax = np.quantile(vals, [0.02, 0.98])

fig, ax = plt.subplots(1, 1, figsize=(14, 9))

gdf_miss = gdf[gdf["county_mean_delay"].isna()]
gdf_has = gdf[gdf["county_mean_delay"].notna()]

gdf_miss.plot(ax=ax, color="0.95", edgecolor="0.85", linewidth=0.2)
norm = Normalize(vmin=float(vmin), vmax=float(vmax))
cmap = plt.get_cmap("viridis")
gdf_has.plot(
    ax=ax,
    column="county_mean_delay",
    cmap=cmap,
    vmin=float(vmin),
    vmax=float(vmax),
    edgecolor="0.85",
    linewidth=0.2,
)

sm = ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])
cbar = fig.colorbar(sm, ax=ax, fraction=0.035, pad=0.02)
cbar.set_label("Mean arrival delay (min)")

airports_gdf.plot(ax=ax, color="black", markersize=45, zorder=6)
for _, r in airports_gdf.iterrows():
    ax.annotate(
        r["Airport"],
        xy=(r.geometry.x, r.geometry.y),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=11,
        fontweight="bold",
    )

half_styles = {
    1000: {"lw": 1.7, "alpha": 0.55},
    2000: {"lw": 2.0, "alpha": 0.40},
    3000: {"lw": 2.3, "alpha": 0.30},
}
for km in HALF_RING_KM:
    r_m = km * 1000.0
    ax.add_patch(
        Arc(
            (midpoint.x, midpoint.y),
            width=2 * r_m,
            height=2 * r_m,
            angle=0,
            theta1=0,
            theta2=180,
            linewidth=half_styles[km]["lw"],
            color="black",
            alpha=half_styles[km]["alpha"],
            zorder=3,
        )
    )

for carrier, mk in carrier_to_marker.items():
    sub = carrier_pts_gdf[carrier_pts_gdf["dominant_carrier"] == carrier]
    if len(sub) == 0:
        continue
    ax.scatter(
        sub.geometry.x,
        sub.geometry.y,
        s=sub["size"],
        marker=mk,
        facecolors="none",
        edgecolors="black",
        linewidths=1.0,
        alpha=0.65,
        zorder=5,
    )

other = carrier_pts_gdf[~carrier_pts_gdf["dominant_carrier"].isin(top_carriers_overall)]
if len(other) > 0:
    ax.scatter(
        other.geometry.x,
        other.geometry.y,
        s=other["size"] * 0.35,
        marker=".",
        color="black",
        alpha=0.22,
        zorder=4,
    )

ax.set_title("County arrival delay (CONUS)", fontsize=13)

minx, miny, maxx, maxy = gdf.total_bounds
pad_x = (maxx - minx) * 0.03
pad_top = (maxy - miny) * 0.05
pad_bottom = (maxy - miny) * 0.01
ax.set_xlim(minx - pad_x, maxx + pad_x)
ax.set_ylim(miny - pad_bottom, maxy + pad_top)

airline_handles = [
    Line2D(
        [0], [0],
        marker=carrier_to_marker[c],
        color="black",
        markerfacecolor="none",
        linestyle="None",
        markersize=11,
        label=c,
    )
    for c in top_carriers_overall
]
airline_handles.append(Line2D([0], [0], marker=".", color="black", linestyle="None", markersize=13, alpha=0.5, label="Other"))

ring_handles = [
    Line2D([0], [0], color="black", linewidth=half_styles[k]["lw"], alpha=half_styles[k]["alpha"], label=f"{k} km")
    for k in HALF_RING_KM
]

plt.subplots_adjust(bottom=0.18)

leg1 = ax.legend(
    handles=airline_handles,
    title="Symbol: Dominant carrier, Size: Frequency",
    loc="upper left",
    bbox_to_anchor=(0.02, -0.08),
    frameon=True,
)
ax.add_artist(leg1)

ax.legend(
    handles=ring_handles,
    title="Distance bands",
    loc="upper right",
    bbox_to_anchor=(0.98, -0.08),
    frameon=True,
)

ax.set_axis_off()
plt.tight_layout()

save_fig(fig, "map_county_choropleth_delay_carrier")
plt.show()

save_text(
    f"airport_key={airport_key}\n"
    f"fips_role={fips_role}\n"
    f"fips_col={fips_col}\n"
    f"airline_col={airline_col}\n"
    f"min_county_flights={MIN_COUNTY_FLIGHTS}\n",
    "map_run_notes.txt",
)
```

# 9/11 and Airport Delay Performance: MCO vs MIA (2001)

## Objective and identification strategy

This analysis evaluates whether the operational shock associated with **9/11** is associated with a **structural change in arrival delay behavior** at two Florida airports:

- **MCO** (Orlando International)
- **MIA** (Miami International)

Two complementary empirical approaches are used:

1. **Flight-level pre/post comparisons (micro evidence)**  
   Compare the distribution of **arrival delays (ArrDelay, minutes)** for completed flights in a defined **pre period** and **post period**, excluding the FAA ground stop window. This assesses whether typical delays and the shape/spread of delays differ before vs after.

2. **Interrupted time series (ITS) / segmented regression (macro evidence)**  
   Aggregate to weekly mean arrival delays and estimate a **segmented regression** with a discrete break at the post-9/11 restart date. This assesses:
   - an **immediate level shift** in weekly mean delays, and
   - a **slope change** (trend change) after the event.

These two views are deliberately complementary: flight-level tests are powerful for detecting distributional change, while ITS provides a clean time-series interpretation of “level and trend” breaks.

---

## Data and measurement

- Outcome: **ArrDelay** in minutes.
- “Completed flights” exclude **Cancelled** and **Diverted** records when analyzing delays (because ArrDelay is not operationally comparable when a flight does not arrive normally).
- The **ground stop period** is excluded from pre/post comparisons to avoid mixing grounded operations with resumed operations.

The analysis also separately evaluates **cancellation rates** (a distinct operational response channel that may rise even when delays fall).

---

## Flight-level results (completed flights, 2001 pre vs post)

### Summary statistics

**MCO**
- Pre (n=71,726): mean **6.28**, median **-2**, p90 **32**, p95 **62**, sd **34.60**
- Post (n=24,328): mean **-2.41**, median **-6**, p90 **13**, p95 **27**, sd **26.31**

**Interpretation (MCO):**  
After 9/11, arrival delays shift downward substantially. Notably, the **upper tail compresses strongly**: p95 falls from **62** to **27** minutes. Dispersion also drops (sd from **34.6** to **26.3**).

**MIA**
- Pre (n=54,863): mean **9.63**, median **0**, p90 **43**, p95 **75.9**, sd **38.52**
- Post (n=18,258): mean **2.70**, median **-4**, p90 **27**, p95 **50**, sd **32.24**

**Interpretation (MIA):**  
MIA also improves post-9/11, with the mean falling by ~**6.9** minutes. The right tail again tightens (p95 from **75.9** to **50**), with reduced dispersion.

### Hypothesis tests (ArrDelay distributions)

For each airport, multiple tests evaluate distinct aspects of change:

- **Mann–Whitney U**: shift in central tendency (nonparametric; robust to skew/heavy tails).
- **Kolmogorov–Smirnov (KS)**: detects any distributional shift (location/scale/shape).
- **Winsorized Welch t-test**: mean shift after limiting extreme tails (robust-ish mean inference).
- **Levene (median-centered)**: variance/dispersion shift.
- **Cliff’s delta**: effect size for stochastic dominance (how often a random pre observation exceeds a random post observation).

**MCO**
- Mann–Whitney p ≈ 0 (very strong evidence of a shift)
- KS D=0.167, p ≈ 0 (distribution shape changed meaningfully)
- Winsorized Welch t=56.6, p ≈ 0 (mean decreased)
- Levene W=500.8, p≈1.21e-110 (variance decreased)
- Cliff’s delta **0.227** (moderate effect size)

**MIA**
- Mann–Whitney p=2.88e-220 (very strong)
- KS D=0.115, p=3.67e-157 (shape/scale shift)
- Winsorized Welch t=29.3, p=1.01e-186 (mean decreased)
- Levene W=155.1, p=1.42e-35 (variance decreased)
- Cliff’s delta **0.158** (small-to-moderate effect size)

**Practical interpretation of Cliff’s delta:**  
A positive delta (PRE vs POST) indicates pre-9/11 delays tend to be larger than post-9/11 delays. MCO shows a larger shift than MIA by this metric.

**MCO**

**MIA**

A key takeaway is that “improved delays” and “worsened cancellations” can occur simultaneously and should be interpreted jointly.

## Interrupted time series (weekly mean ArrDelay, extended window)

### ITS model and coefficients

The weekly segmented regression is:

$
\text{ArrDelay}_t = \beta_0 + \beta_1 t + \beta_2 \text{post}_t + \beta_3 (t \times \text{post})_t + \varepsilon_t
$

Where:
- $t$ = week index (time trend)
- $\text{post}$ = 1 after operational restart date, 0 before
- $t\_post$ = time since restart (0 before post, increasing after)

Inference uses **HAC (Newey–West)** standard errors to reduce sensitivity to autocorrelation in weekly series.

### Airport-specific ITS results

**MCO**
- Level shift (post): **-7.53** minutes (p < 0.001)
- Pre-trend slope (t): -0.036 (p = 0.186) not significant
- Post-trend change (t_post): +0.066 (p = 0.095) marginal
- Joint test (post = 0 and t_post = 0): chi2=21.63, p=2.01e-05

**Interpretation (MCO):**  
There is strong evidence of an **immediate downward shift** in weekly mean arrival delays (~7.5 minutes). Evidence for a trend change is weaker (marginal), suggesting the dominant effect is a **level drop** rather than a major change in slope.

**MIA**
- Level shift (post): **-8.23** minutes (p < 0.001)
- Pre-trend slope (t): +0.016 (p = 0.551) not significant
- Post-trend change (t_post): -0.017 (p = 0.657) not significant
- Joint test: chi2=26.50, p=1.76e-06

**Interpretation (MIA):**  
MIA shows a strong **immediate downward shift** (~8.2 minutes) with **no evidence** of a meaningful slope change afterward. Again, the effect is primarily a level shift.

### Pooled weekly model (difference in changes across airports)

The pooled model includes airport and airport×post, airport×t_post interactions (MCO is the reference).

Key results:
- Baseline airport difference (MIA vs MCO): +2.22 minutes (p = 0.063) marginal
- Common post level shift: **-8.71** minutes (p < 0.001)
- Additional post shift for MIA vs MCO: +1.66 (p = 0.332) not significant
- Additional slope change for MIA vs MCO: -0.032 (p = 0.440) not significant

**Interpretation:**  
Both airports experience a similar **post-9/11 drop** in weekly mean arrival delays. The data do not support a statistically distinct “MIA response” relative to MCO in either level shift or slope change in the weekly series.

  - lower central tendency, and
  - less week-to-week volatility (consistent with reduced variance in flight-level tests).

## Synthesis and operational interpretation

### Main empirical conclusion
Across both micro (flight-level) and macro (ITS) evidence, the post-9/11 period is associated with:

1. **Lower arrival delays for completed flights** at both MCO and MIA  
   - Mean ArrDelay falls by roughly **8–9 minutes at MCO** and **~7 minutes at MIA** in the 2001 pre/post window.
   - Upper-tail delays compress materially (large drops in p90/p95).

2. **Higher cancellation rates** at both airports  
   - Cancellations roughly **double** (MCO: ~1.5%→3.1%, MIA: ~2.0%→3.9%).

### Why “delays down” does not contradict “system disruption”
This combination is consistent with a plausible mechanism:
- **Demand/capacity contraction** and intensified security procedures changed the operating environment.
- Airlines likely reduced schedules, and more flights were canceled (especially during recovery/uncertainty).
- With fewer aircraft movements, congestion and queueing pressures can fall, leading to **better delay performance among flights that do operate**.

Thus, 9/11 appears to shift the system toward:
- fewer operated flights (more cancellations),
- but improved on-time performance conditional on operating.

---

## Bottom line
For both **MCO** and **MIA**, the post-9/11 period shows a statistically and practically significant **structural reduction in arrival delays among completed flights**, accompanied by a significant **increase in cancellations**. The ITS results indicate that the dominant change is an **immediate level shift** rather than a sustained change in trend, and pooled weekly results suggest the magnitude of the break is broadly similar across the two airports.


```python
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf

# -----------------------------
# Config
# -----------------------------
OUTDIR = globals().get("OUTDIR", "../visualizations/2_Delay_Deep_Dive")
os.makedirs(OUTDIR, exist_ok=True)

BLUE = "#1f77b4"  # MCO
RED = "#d62728"   # MIA
AIRPORT_COLORS = {"MCO": BLUE, "MIA": RED}

DATA_PATH = "../data/processed/mco_mia_clean_long.csv.gz"

EVENT_DATE = pd.Timestamp("2001-09-11")
GROUND_START = pd.Timestamp("2001-09-11")
GROUND_END = pd.Timestamp("2001-09-13")

PRE_START = pd.Timestamp("2001-01-01")
PRE_END = pd.Timestamp("2001-09-10")
POST_START = pd.Timestamp("2001-09-14")
POST_END = pd.Timestamp("2001-12-31")

EXT_START = PRE_START - pd.DateOffset(years=1)
EXT_END = POST_END + pd.DateOffset(years=1)

DESIRED_COLS = [
    "FlightDate",
    "ArrDelay", "DepDelay",
    "Cancelled", "Diverted",
    "UniqueCarrier", "DayOfWeek", "TimeOfDay",
    "Airport", "Origin", "Dest",
]

# -----------------------------
# Warnings (keep tight)
# -----------------------------
warnings.filterwarnings(
    "ignore",
    message="The behavior of wald_test will change after 0.14",
    category=FutureWarning,
)
warnings.filterwarnings(
    "ignore",
    message="Conversion of an array with ndim > 0 to a scalar is deprecated",
    category=DeprecationWarning,
)

# -----------------------------
# IO helpers
# -----------------------------
def save_fig(fig, stem: str, dpi: int = 200):
    png = os.path.join(OUTDIR, f"{stem}.png")
    pdf = os.path.join(OUTDIR, f"{stem}.pdf")
    fig.savefig(png, dpi=dpi, bbox_inches="tight")
    fig.savefig(pdf, bbox_inches="tight")
    return png, pdf

def save_df(df: pd.DataFrame, filename: str):
    path = os.path.join(OUTDIR, filename)
    df.to_csv(path, index=False)
    return path

def save_text(text: str, filename: str):
    path = os.path.join(OUTDIR, filename)
    with open(path, "w", encoding="utf-8") as f:
        f.write(text)
    return path

# -----------------------------
# ETL (chunked)
# -----------------------------
def _detect_usecols(path: str, desired_cols):
    hdr = pd.read_csv(path, compression="gzip", nrows=0)
    cols = list(hdr.columns)
    usecols = [c for c in desired_cols if c in cols]
    missing_core = [c for c in ["FlightDate", "ArrDelay", "DepDelay", "Cancelled", "Diverted"] if c not in usecols]
    if missing_core:
        raise ValueError(
            f"Missing required columns: {missing_core}. "
            f"Available: {cols[:50]}{'...' if len(cols) > 50 else ''}"
        )
    return usecols, cols

def _infer_airport_column(all_columns):
    for c in ["Airport", "Origin", "Dest"]:
        if c in all_columns:
            return c
    return None

def load_filtered(path: str, chunksize: int = 400_000, lo: pd.Timestamp = None, hi: pd.Timestamp = None) -> pd.DataFrame:
    lo = PRE_START if lo is None else lo
    hi = POST_END if hi is None else hi

    usecols, all_cols = _detect_usecols(path, DESIRED_COLS)
    airport_src_col = _infer_airport_column(all_cols)
    if airport_src_col is None:
        raise ValueError("No airport column found (expected Airport/Origin/Dest).")

    out = []
    reader = pd.read_csv(
        path,
        compression="gzip",
        usecols=usecols,
        chunksize=chunksize,
        low_memory=True,
    )

    for ch in reader:
        ch["FlightDate"] = pd.to_datetime(ch["FlightDate"], errors="coerce")
        ch = ch[(ch["FlightDate"] >= lo) & (ch["FlightDate"] <= hi)].copy()
        if ch.empty:
            continue

        for b in ["Cancelled", "Diverted"]:
            if b in ch.columns:
                ch[b] = ch[b].astype("boolean").fillna(False).astype(bool)

        ch["airport"] = ch[airport_src_col].astype(str).str.upper()
        out.append(ch)

    if not out:
        raise ValueError("No rows found in the requested date range.")

    df = pd.concat(out, ignore_index=True)
    df = df[df["airport"].isin(["MCO", "MIA"])].copy()
    if df.empty:
        raise ValueError("No rows remain for MCO/MIA after filtering.")
    return df

df_all = load_filtered(DATA_PATH, lo=EXT_START, hi=EXT_END)

df_all["period"] = np.where(
    (df_all["FlightDate"] >= PRE_START) & (df_all["FlightDate"] <= PRE_END),
    "pre",
    np.where(
        (df_all["FlightDate"] >= POST_START) & (df_all["FlightDate"] <= POST_END),
        "post",
        "other",
    ),
)

completed_all = df_all[(~df_all["Cancelled"]) & (~df_all["Diverted"])].copy()
completed_all = completed_all.dropna(subset=["DepDelay", "ArrDelay"]).copy()

save_df(df_all[["FlightDate", "airport", "ArrDelay", "DepDelay", "Cancelled", "Diverted", "period"]], "911_filtered_flights.csv")

# -----------------------------
# Stats: flight-level pre/post (2001)
# -----------------------------
def summarize_delay(x: np.ndarray) -> dict:
    x = np.asarray(x)
    return {
        "n": int(x.size),
        "mean": float(np.mean(x)) if x.size else np.nan,
        "median": float(np.median(x)) if x.size else np.nan,
        "p90": float(np.quantile(x, 0.90)) if x.size else np.nan,
        "p95": float(np.quantile(x, 0.95)) if x.size else np.nan,
        "std": float(np.std(x, ddof=1)) if x.size > 1 else np.nan,
    }

def winsorize(x: np.ndarray, p: float = 0.01) -> np.ndarray:
    x = np.asarray(x)
    if x.size == 0:
        return x
    lo, hi = np.quantile(x, [p, 1 - p])
    return np.clip(x, lo, hi)

def cliffs_delta(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x)
    y = np.asarray(y)
    if x.size == 0 or y.size == 0:
        return np.nan

    max_n = 20000
    if x.size > max_n:
        x = np.random.default_rng(0).choice(x, size=max_n, replace=False)
    if y.size > max_n:
        y = np.random.default_rng(1).choice(y, size=max_n, replace=False)

    if x.size * y.size > 20_000_000:
        pooled = np.concatenate([x, y])
        ranks = stats.rankdata(pooled)
        rx = ranks[:x.size].sum()
        U = rx - x.size * (x.size + 1) / 2
        return float((2 * U) / (x.size * y.size) - 1)

    gt = (x[:, None] > y[None, :]).sum()
    lt = (x[:, None] < y[None, :]).sum()
    return float((gt - lt) / (x.size * y.size))

airports = ["MCO", "MIA"]
completed_2001 = completed_all[completed_all["period"].isin(["pre", "post"])].copy()

rows = []
notes = []
for ap in airports:
    sub = completed_2001[completed_2001["airport"] == ap]

    pre_arr = sub.loc[sub["period"] == "pre", "ArrDelay"].to_numpy()
    post_arr = sub.loc[sub["period"] == "post", "ArrDelay"].to_numpy()

    mw = stats.mannwhitneyu(pre_arr, post_arr, alternative="two-sided")
    ks = stats.ks_2samp(pre_arr, post_arr)
    tt = stats.ttest_ind(winsorize(pre_arr), winsorize(post_arr), equal_var=False, nan_policy="omit")
    lv = stats.levene(pre_arr, post_arr, center="median")
    cd = cliffs_delta(pre_arr, post_arr)

    rows.append({
        "airport": ap,
        "pre_n": int(pre_arr.size),
        "post_n": int(post_arr.size),
        "pre_mean": summarize_delay(pre_arr)["mean"],
        "post_mean": summarize_delay(post_arr)["mean"],
        "mw_U": float(mw.statistic),
        "mw_p": float(mw.pvalue),
        "ks_D": float(ks.statistic),
        "ks_p": float(ks.pvalue),
        "welch_t": float(tt.statistic),
        "welch_p": float(tt.pvalue),
        "levene_W": float(lv.statistic),
        "levene_p": float(lv.pvalue),
        "cliffs_delta": float(cd),
    })

    notes.append(f"[{ap}] ArrDelay pre={summarize_delay(pre_arr)} post={summarize_delay(post_arr)}")

flight_tests = pd.DataFrame(rows)
save_df(flight_tests, "911_flight_level_tests_arrdelay.csv")
save_text("\n".join(notes) + "\n", "911_flight_level_summaries.txt")

# -----------------------------
# Stats: cancellation rate pre/post (2001)
# -----------------------------
df_2001 = df_all[df_all["period"].isin(["pre", "post"])].copy()

rows = []
for ap in airports:
    sub = df_2001[df_2001["airport"] == ap]
    pre_cancel = float(sub.loc[sub["period"] == "pre", "Cancelled"].mean())
    post_cancel = float(sub.loc[sub["period"] == "post", "Cancelled"].mean())
    n_pre = int((sub["period"] == "pre").sum())
    n_post = int((sub["period"] == "post").sum())

    count = np.array([pre_cancel * n_pre, post_cancel * n_post], dtype=float)
    nobs = np.array([n_pre, n_post], dtype=float)
    zstat, pval = sm.stats.proportions_ztest(count=count, nobs=nobs)

    rows.append({
        "airport": ap,
        "pre_cancel_rate": pre_cancel,
        "post_cancel_rate": post_cancel,
        "n_pre": n_pre,
        "n_post": n_post,
        "z": float(zstat),
        "p": float(pval),
    })

cancel_tests = pd.DataFrame(rows)
save_df(cancel_tests, "911_cancellation_rate_tests.csv")

# -----------------------------
# ITS weekly aggregation + models
# -----------------------------
def make_weekly(completed_df: pd.DataFrame) -> pd.DataFrame:
    tmp = completed_df.copy()
    tmp["week"] = tmp["FlightDate"].dt.to_period("W").apply(lambda r: r.start_time)

    w = (
        tmp.groupby(["airport", "week"], as_index=False)
        .agg(
            arr_mean=("ArrDelay", "mean"),
            arr_sem=("ArrDelay", lambda s: s.std(ddof=1) / np.sqrt(len(s)) if len(s) > 1 else np.nan),
            n=("ArrDelay", "size"),
        )
        .sort_values(["airport", "week"])
        .copy()
    )

    w["t"] = w.groupby("airport")["week"].transform(lambda x: (x - x.min()).dt.days / 7.0)
    w["post"] = (w["week"] >= POST_START).astype(int)

    w["t_post"] = w["t"] - w.groupby("airport")["t"].transform(
        lambda s: s[w.loc[s.index, "post"] == 1].min()
    )
    w.loc[w["post"] == 0, "t_post"] = 0.0
    return w

completed_plot1 = completed_all[(completed_all["FlightDate"] >= PRE_START) & (completed_all["FlightDate"] <= POST_END)].copy()
weekly_2001 = make_weekly(completed_plot1)

completed_plot2 = completed_all[(completed_all["FlightDate"] >= EXT_START) & (completed_all["FlightDate"] <= EXT_END)].copy()
weekly_ext = make_weekly(completed_plot2)

def _to_scalar(x):
    arr = np.asarray(x)
    return float(arr.reshape(-1)[0])

its_models_arr = {}
its_rows = []
for ap in airports:
    w = weekly_ext[weekly_ext["airport"] == ap].copy()
    mod = smf.ols("arr_mean ~ t + post + t_post", data=w).fit(cov_type="HAC", cov_kwds={"maxlags": 4})
    its_models_arr[ap] = mod

    wald = mod.wald_test("post = 0, t_post = 0", scalar=True)

    its_rows.append({
        "airport": ap,
        "post_coef": float(mod.params.get("post", np.nan)),
        "t_post_coef": float(mod.params.get("t_post", np.nan)),
        "wald_chi2": _to_scalar(wald.statistic),
        "wald_p": _to_scalar(wald.pvalue),
        "n_weeks": int(w.shape[0]),
    })

    save_text(str(mod.summary()), f"911_its_model_{ap}.txt")

its_summary = pd.DataFrame(its_rows)
save_df(its_summary, "911_its_summary_by_airport.csv")

weekly_ext["airport"] = pd.Categorical(weekly_ext["airport"], categories=["MCO", "MIA"], ordered=True)
pooled_arr = smf.ols(
    "arr_mean ~ t + post + t_post + C(airport) + C(airport):post + C(airport):t_post",
    data=weekly_ext,
).fit(cov_type="HAC", cov_kwds={"maxlags": 4})
save_text(str(pooled_arr.summary()), "911_weekly_pooled_model.txt")

# -----------------------------
# Figure 1 (2001 window): weekly mean + 95% CI
# -----------------------------
z = 1.96
weekly_2001["arr_ci_lo"] = weekly_2001["arr_mean"] - z * weekly_2001["arr_sem"]
weekly_2001["arr_ci_hi"] = weekly_2001["arr_mean"] + z * weekly_2001["arr_sem"]

fig1, ax1 = plt.subplots(figsize=(12, 6))

for ap in airports:
    w = weekly_2001[weekly_2001["airport"] == ap]
    ax1.plot(w["week"], w["arr_mean"], color=AIRPORT_COLORS[ap], linewidth=2, label=ap)
    ax1.fill_between(w["week"], w["arr_ci_lo"], w["arr_ci_hi"], color=AIRPORT_COLORS[ap], alpha=0.15)

ax1.axvline(EVENT_DATE, linestyle="--", linewidth=1)
ax1.axvspan(GROUND_START, GROUND_END, alpha=0.12)

ymin, ymax = ax1.get_ylim()
ax1.text(EVENT_DATE, ymax, "  9/11", va="top")
ax1.text(GROUND_START, ymin, "Ground stop", va="bottom")

ax1.set_title("Weekly mean arrival delay (2001)")
ax1.set_xlabel(None)
ax1.set_ylabel("min")
ax1.grid(True, alpha=0.25)
ax1.legend(ncol=2, frameon=False)

fig1.tight_layout()
save_fig(fig1, "911_fig1_weekly_mean_ci_2001")
plt.show()

# -----------------------------
# Figure 2 (extended window): weekly mean + fitted segmented trend
# -----------------------------
fig2, ax2 = plt.subplots(figsize=(12, 6))

for ap in airports:
    w = weekly_ext[weekly_ext["airport"] == ap].copy()
    mod = its_models_arr[ap]
    w["arr_fit"] = mod.predict(w)

    ax2.plot(w["week"], w["arr_mean"], color=AIRPORT_COLORS[ap], linewidth=1.6, alpha=0.55)
    ax2.plot(w["week"], w["arr_fit"], color=AIRPORT_COLORS[ap], linewidth=3, label=ap)

ax2.axvline(EVENT_DATE, linestyle="--", linewidth=1)
ax2.axvspan(GROUND_START, GROUND_END, alpha=0.12)

ymin, ymax = ax2.get_ylim()
ax2.text(EVENT_DATE, ymax, "  9/11", va="top")
ax2.text(GROUND_START, ymin, "Ground stop", va="bottom")

ax2.set_title("Arrival delay level/trend shift (segmented fit)")
ax2.set_xlabel(None)
ax2.set_ylabel("min")
ax2.grid(True, alpha=0.25)
ax2.legend(ncol=2, frameon=False)

fig2.tight_layout()
save_fig(fig2, "911_fig2_weekly_mean_and_fit_extended")
plt.show()

save_text("DONE (9/11 analysis). Outputs saved to OUTDIR.\n", "911_run_log.txt")
```
