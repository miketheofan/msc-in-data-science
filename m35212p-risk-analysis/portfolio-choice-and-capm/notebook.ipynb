## 2. Return Construction and Descriptive Statistics

We compute **weekly simple returns** for every asset, split the sample into the estimation and

## 3. Diversification Analysis

Two simple portfolios: (1) **equal-weight** all 10 stocks; (2) **concentrated** in the 2 most volatile
stocks. For each: annualized return, annualized volatility, Sharpe ratio and cumulative performance.
The reusable portfolio helpers are defined here and used again in Sections 4–5.

````python
# --- Reusable portfolio analytics (annualized) ---
mean_weekly = ret_est_stocks.mean()   # expected-return estimate used throughout

def port_return(weights, mean_ret=mean_weekly, ann=WEEKS_PER_YEAR):
    # Annualized portfolio return from weekly mean returns
    return float(np.dot(weights, mean_ret) * ann)

def port_vol(weights, cov=cov_matrix, ann=WEEKS_PER_YEAR):
    # Annualized portfolio volatility from the weekly covariance matrix
    w = np.asarray(weights)
    return float(np.sqrt(w @ cov.values @ w) * np.sqrt(ann))

def port_sharpe(weights, mean_ret=mean_weekly, cov=cov_matrix, rf=RF_ANNUAL):
    return (port_return(weights, mean_ret) - rf) / port_vol(weights, cov)

def max_drawdown_est(weights):
    # Maximum drawdown over the estimation period
    r = ret_est_stocks @ np.asarray(weights)
    cum = (1 + r).cumprod()
    running_max = cum.cummax()
    return float((cum / running_max - 1.0).min())

# --- Portfolio 1: equally weighted across all 10 stocks ---
w_eq = pd.Series(1.0 / len(STOCKS), index=STOCKS)

# --- Portfolio 2: concentrated in the 2 most volatile stocks (50/50) ---
two_most_vol = desc["Ann. vol"].sort_values(ascending=False).head(2).index.tolist()
w_conc = pd.Series(0.0, index=STOCKS)
w_conc[two_most_vol] = 0.5
print("Two most volatile stocks:", two_most_vol)

# --- Summary metrics for both portfolios (including max drawdown) ---
simple_summary = pd.DataFrame({
    "Equal-weight (10)": [port_return(w_eq), port_vol(w_eq), port_sharpe(w_eq), max_drawdown_est(w_eq)],
    f"Concentrated ({' + '.join(two_most_vol)})": [port_return(w_conc), port_vol(w_conc), port_sharpe(w_conc), max_drawdown_est(w_conc)],
}, index=["Ann. return", "Ann. volatility", "Sharpe ratio", "Max drawdown"]).round(4)
display(simple_summary)

# --- Cumulative performance over the estimation period ---
cum_eq   = (1 + ret_est_stocks @ w_eq.values).cumprod()
cum_conc = (1 + ret_est_stocks @ w_conc.values).cumprod()

fig, ax = plt.subplots()
cum_eq.plot(ax=ax, label="Equal-weight (10 stocks)")
cum_conc.plot(ax=ax, label=f"Concentrated ({' + '.join(two_most_vol)})")
ax.set_title("Cumulative performance - estimation period")
ax.set_ylabel("Growth of $1")
ax.legend()
plt.tight_layout()
plt.show()
````

### Interpretations 

1. The equal-weight portfolio has an annualized volatility of approximately 20.1% with a Sharpe ratio of 1.21, while the concentrated FCX and LRCX portfolio earns a higher return (39%) but at 39.7% volatility and a lower Sharpe ratio of 0.99. Crucially, the 10-stock volatility (20.1%) sits far below either of its most volatile constituents (FCX at 53.6%, LRCX at 40.8%) and also below their simple 50/50 average. This gap represents the diversification benefit.

2. Holding more stocks does not automatically reduce risk. The reduction comes from combining assets with low pairwise correlations, not from the count itself. FCX and LRCX have a correlation of approximately 0.40, so combining them barely diversifies and actually produces the worst Sharpe ratio. Spreading the same capital across ten weakly correlated sectors is what cuts portfolio variance roughly in half relative to the concentrated portfolio.

3. The maximum drawdown comparison reinforces this point. During the estimation period, the concentrated portfolio suffered a peak-to-trough loss of approximately 47%, meaning an investor would have seen nearly half their wealth evaporate at the worst point. The equal-weight portfolio experienced a drawdown of roughly 28%, which is still painful but far more survivable. The concentrated portfolio earned higher total returns over the full period, but those gains came with the real possibility of catastrophic interim losses. An investor who needed liquidity during a drawdown, or who lost confidence and sold near the bottom, would have locked in large losses that the eventual recovery could not undo. This illustrates why risk-adjusted measures like the Sharpe ratio matter more than raw returns when evaluating portfolio quality.

## 4. Efficient Frontier Without Short-Selling

Mean–variance frontier with $w_i \ge 0$ and $\sum_i w_i = 1$. Solve for the minimum-variance and
maximum-Sharpe (tangency) portfolios, trace the frontier, and mark the individual stocks, the
equal-weight, minimum-variance and tangency portfolios. Report the MVP and tangency weights.

````python
N = len(STOCKS)
bounds_long = tuple((0.0, 1.0) for _ in range(N))   # no short-selling
budget = {"type": "eq", "fun": lambda w: np.sum(w) - 1.0}
w0 = np.repeat(1.0 / N, N)

# --- Minimum-variance portfolio (long-only) ---
res_mvp = minimize(port_vol, w0, method="SLSQP", bounds=bounds_long, constraints=[budget])
w_mvp = pd.Series(res_mvp.x, index=STOCKS)

# --- Tangency portfolio = maximum Sharpe ratio (long-only) ---
neg_sharpe = lambda w: -port_sharpe(w)
res_tan = minimize(neg_sharpe, w0, method="SLSQP", bounds=bounds_long, constraints=[budget])
w_tan = pd.Series(res_tan.x, index=STOCKS)

# --- Trace the efficient frontier: minimise vol for a grid of target returns ---
def frontier(bounds, n=60):
    ann_mean = mean_weekly * WEEKS_PER_YEAR
    targets = np.linspace(ann_mean.min(), ann_mean.max(), n)
    vols, rets = [], []
    for t in targets:
        cons = [budget, {"type": "eq", "fun": lambda w, t=t: port_return(w) - t}]
        r = minimize(port_vol, w0, method="SLSQP", bounds=bounds, constraints=cons)
        if r.success:
            vols.append(port_vol(r.x)); rets.append(t)
    return np.array(vols), np.array(rets)

fr_vol, fr_ret = frontier(bounds_long)

# --- Plot frontier with individual stocks and key portfolios ---
fig, ax = plt.subplots()
ax.plot(fr_vol, fr_ret, "b-", lw=2, label="Efficient frontier (no short)")
ax.scatter(desc["Ann. vol"], desc["Ann. mean"], c="grey", marker="o", zorder=3)
for s in STOCKS:
    ax.annotate(s, (desc.loc[s, "Ann. vol"], desc.loc[s, "Ann. mean"]),
                fontsize=8, xytext=(4, 2), textcoords="offset points")
ax.scatter(port_vol(w_eq),  port_return(w_eq),  c="green",  marker="s", s=90,  label="Equal-weight", zorder=4)
ax.scatter(port_vol(w_mvp), port_return(w_mvp), c="purple", marker="*", s=220, label="Min-variance", zorder=4)
ax.scatter(port_vol(w_tan), port_return(w_tan), c="red",    marker="*", s=220, label="Tangency", zorder=4)
ax.set_xlabel("Annualized volatility"); ax.set_ylabel("Annualized return")
ax.set_title("Efficient frontier without short-selling")
ax.legend(); plt.tight_layout(); plt.show()

# --- Key portfolio statistics ---
print(f"Min-variance : ret={port_return(w_mvp):.4f}  vol={port_vol(w_mvp):.4f}  Sharpe={port_sharpe(w_mvp):.4f}")
print(f"Tangency     : ret={port_return(w_tan):.4f}  vol={port_vol(w_tan):.4f}  Sharpe={port_sharpe(w_tan):.4f}")

# --- Report weights (drop near-zero rows for readability) ---
weights_long = pd.DataFrame({"Min-variance": w_mvp, "Tangency": w_tan})
display(weights_long[weights_long.abs().max(axis=1) > 1e-4].round(4))
````

### Interpretations 

1. Both optimized portfolios load most heavily on the low-volatility, low-correlation defensive stocks. COST receives a weight of 0.32 in the minimum-variance portfolio and 0.30 in the tangency portfolio, with NEE, DE and TTWO also receiving significant allocations. The tangency portfolio additionally takes a meaningful LRCX stake (0.14) to capture its high average return, lifting its Sharpe ratio to 1.45 (return 24.6%, volatility 17.0%) compared to the minimum-variance portfolio's Sharpe of 1.34 (return 20.8%, volatility 15.5%).

2. These weights are economically intuitive. The minimum-variance portfolio tilts toward the steadiest, least correlated names. Several assets are excluded with weights near zero. FCX and LULU are excluded from both portfolios, while EOG and SPGI are excluded from the tangency portfolio. FCX is dropped despite its 37% return because its 53.6% volatility and high covariance with EOG, DE and LRCX make it inefficient on a risk-adjusted basis. This corner-solution behaviour, where the optimizer concentrates in a handful of names and zeros out the rest, is a well-known feature of mean-variance optimization under a no-short-selling constraint.

3. Looking at the frontier plot, we observe that all individual stocks lie to the right of or below the efficient frontier curve, confirming that no single stock is efficient on its own. The equal-weight portfolio sits inside the frontier, meaning it is suboptimal because we could achieve the same return with less risk or more return with the same risk. Both the minimum-variance and tangency portfolios lie directly on the frontier, representing the best achievable combinations under the given constraints. The tangency portfolio sits higher and slightly to the right of the minimum-variance portfolio, reflecting the tradeoff between risk and return along the efficient set.

## 5. Efficient Frontier With Bounded Short-Selling

Repeat with $-1 \le w_i \le 1$ and $\sum_i w_i = 1$. Overlay the no-short and bounded-short frontiers
on one graph.

````python
bounds_short = tuple((-1.0, 1.0) for _ in range(N))   # bounded short-selling

# --- Tangency portfolio with bounded short-selling ---
res_tan_s = minimize(neg_sharpe, w0, method="SLSQP", bounds=bounds_short, constraints=[budget])
w_tan_short = pd.Series(res_tan_s.x, index=STOCKS)

# --- Bounded-short frontier ---
frs_vol, frs_ret = frontier(bounds_short)

# --- Overlay both frontiers ---
fig, ax = plt.subplots()
ax.plot(fr_vol, fr_ret, "b-", lw=2, label="No short-selling (0 <= w <= 1)")
ax.plot(frs_vol, frs_ret, "r--", lw=2, label="Bounded short-selling (-1 <= w <= 1)")
ax.scatter(desc["Ann. vol"], desc["Ann. mean"], c="grey", marker="o", zorder=3)
for s in STOCKS:
    ax.annotate(s, (desc.loc[s, "Ann. vol"], desc.loc[s, "Ann. mean"]),
                fontsize=8, xytext=(4, 2), textcoords="offset points")
ax.set_xlabel("Annualized volatility"); ax.set_ylabel("Annualized return")
ax.set_title("Efficient frontier: no-short vs bounded short-selling")
ax.legend(); plt.tight_layout(); plt.show()

print(f"Tangency (bounded short): ret={port_return(w_tan_short):.4f}  "
      f"vol={port_vol(w_tan_short):.4f}  Sharpe={port_sharpe(w_tan_short):.4f}")

# --- Compare tangency weights under the two regimes ---
tan_compare = pd.DataFrame({"Tangency (long-only)": w_tan,
                            "Tangency (bounded short)": w_tan_short}).round(4)
display(tan_compare)
````

### Interpretations

1. Allowing bounded short-selling shifts the frontier only slightly upward and to the left, and the tangency Sharpe ratio improves marginally from 1.45 to 1.47 (return 25.5%, volatility 17.4%). The optimizer takes modest short positions in LULU (approximately -0.05), EOG (-0.06) and SPGI (-0.04), while amplifying the existing long positions in COST (0.33) and DE (0.23). The tangency portfolio becomes only mildly more extreme because the long-only solution was already close to optimal for this universe.

2. Short-selling improves the mathematical frontier because it relaxes the non-negativity constraint on weights, expanding the feasible set. However, it raises practical investment risk. Short positions carry leverage and theoretically unbounded losses, incur borrowing and financing costs, require margin, and most importantly they magnify estimation error. The weights are built from noisy historical means and covariances, so the extra flexibility tends to overfit the in-sample data and can hurt performance out of sample, as Section 7 illustrates.

3. In plain terms, allowing short-selling gives the optimizer more tools to work with, but the improvement here is minimal. The red dashed frontier barely moves compared to the blue solid one. This tells us that for this particular set of stocks, the long-only constraint was not very restrictive. The stocks the optimizer wants to short (LULU, EOG, SPGI) were already excluded or minimized in the long-only solution. Giving the optimizer permission to bet against them only marginally improves performance, while introducing additional practical risks that may not be worth the small gain.

## 6. CAPM and Beta Estimation

With $R_f=0$, regress each stock on the market: $R_{i,t}=\alpha_i+\beta_i R_{M,t}+\varepsilon_{i,t}$
(OLS via `statsmodels`). Report beta, alpha, realized annual return, CAPM-implied return and the
difference. Compute portfolio betas (EW, MVP, tangency) and draw the Security Market Line.

````python
# --- Per-stock CAPM regressions: R_i = alpha + beta * R_m   (Rf = 0) ---
X = sm.add_constant(ret_est_mkt)                 # market returns + intercept
mkt_ann_return = ret_est_mkt.mean() * WEEKS_PER_YEAR

rows, betas = {}, {}
for s in STOCKS:
    model = sm.OLS(ret_est_stocks[s], X).fit()
    beta  = model.params[MARKET]
    betas[s] = beta
    realized     = mean_weekly[s] * WEEKS_PER_YEAR
    capm_implied = RF_ANNUAL + beta * (mkt_ann_return - RF_ANNUAL)
    rows[s] = {
        "Beta": beta,
        "Alpha (ann.)": model.params["const"] * WEEKS_PER_YEAR,
        "Realized ann. ret": realized,
        "CAPM-implied ret": capm_implied,
        "Diff (realized - CAPM)": realized - capm_implied,
    }
capm_table = pd.DataFrame(rows).T.round(4)
betas = pd.Series(betas)
display(capm_table)

# --- Portfolio betas = weighted average of stock betas ---
port_betas = pd.Series({
    "Equal-weight": float(betas @ w_eq.values),
    "Min-variance": float(betas @ w_mvp.values),
    "Tangency":     float(betas @ w_tan.values),
}).round(4)
print("Portfolio betas:")
print(port_betas)

# --- Security Market Line ---
fig, ax = plt.subplots()
ax.scatter(betas, capm_table["Realized ann. ret"], c="steelblue", zorder=3)
for s in STOCKS:
    ax.annotate(s, (betas[s], capm_table.loc[s, "Realized ann. ret"]),
                fontsize=8, xytext=(4, 2), textcoords="offset points")
bx = np.linspace(0, betas.max() * 1.1, 50)
ax.plot(bx, RF_ANNUAL + bx * (mkt_ann_return - RF_ANNUAL), "r--", label="CAPM / SML")
ax.scatter(1.0, mkt_ann_return, c="black", marker="D", s=80, label="Market (SPY)", zorder=4)
ax.set_xlabel("Beta"); ax.set_ylabel("Average annualized return")
ax.set_title("Security Market Line"); ax.legend()
plt.tight_layout(); plt.show()
````

### Interpretations

1. Market exposure lines up with sector risk as expected. The highest beta stocks are FCX (1.94), LRCX (1.66) and LULU (1.31), while the lowest beta stocks are NEE (0.62), TTWO (0.72), COST (0.72) and VRTX (0.73). The tangency portfolio has a moderate-to-low beta of approximately 0.88 and the minimum-variance portfolio has an even lower beta of 0.74. Both are defensive relative to the market, consistent with their tilt toward low-volatility names.

2. Do high-beta stocks earn more? The relationship is loose at best. The two highest-beta names (LRCX and FCX) did post top returns, but LULU (beta of 1.31) and EOG (beta of 1.21) underperformed the much lower-beta TTWO and COST. The beta-return relationship is therefore noisy rather than monotonic. Moreover, almost every stock shows a positive alpha (only EOG is negative) and plots above the Security Market Line, meaning realized returns generally exceeded the CAPM prediction. The evidence therefore gives the CAPM weak and partial support. Beta captures some of the cross-sectional spread in returns, but the large positive alphas, which are an in-sample artifact that should not be expected to persist, show that beta alone does not fully explain returns in this sample.

3. Looking at the Security Market Line plot, we see that most stocks sit above the red dashed line rather than on it. If the CAPM held perfectly, all stocks would lie exactly on the line, meaning their returns would be fully explained by their beta. Instead, the scatter suggests that other factors beyond market exposure contributed to returns during this period. The optimized portfolios ended up with low betas because they favour defensive stocks, which means they are less sensitive to market swings. This defensive positioning is a natural outcome of the optimization process rather than an intentional bet against the market.

## 7. Out-of-Sample Evaluation

Freeze the weights estimated as of 2025-12-31 and apply them to the out-of-sample window
(2026-Q1): equal-weight, minimum-variance, tangency, and bounded-short tangency. For each:
cumulative return, annualized vol, Sharpe, max drawdown, OOS beta and Jensen's alpha. Plot all
trajectories against SPY.

````python
def max_drawdown(cum_series):
    # cum_series: cumulative wealth index (starts at 1). Worst peak-to-trough drop.
    running_max = cum_series.cummax()
    return float((cum_series / running_max - 1.0).min())

def evaluate_oos(weights, name):
    # Weekly out-of-sample portfolio returns from frozen weights
    r = ret_oos_stocks @ np.asarray(weights)
    cum = (1 + r).cumprod()
    capm = sm.OLS(r, sm.add_constant(ret_oos_mkt)).fit()   # OOS beta + Jensen's alpha
    return pd.Series({
        "Cumulative return":   cum.iloc[-1] - 1,
        "Ann. vol":            r.std() * np.sqrt(WEEKS_PER_YEAR),
        "Sharpe":              (r.mean() * WEEKS_PER_YEAR - RF_ANNUAL) / (r.std() * np.sqrt(WEEKS_PER_YEAR)),
        "Max drawdown":        max_drawdown(cum),
        "Beta (OOS)":          capm.params[MARKET],
        "Jensen alpha (ann.)": capm.params["const"] * WEEKS_PER_YEAR,
    }, name=name), cum

# Freeze the estimation-period weights and apply them out of sample
portfolios = {
    "Equal-weight":             w_eq.values,
    "Min-variance":             w_mvp.values,
    "Tangency":                 w_tan.values,
    "Tangency (bounded short)": w_tan_short.values,
}

results, cums = [], {}
for name, w in portfolios.items():
    s, cum = evaluate_oos(w, name)
    results.append(s); cums[name] = cum
oos_table = pd.DataFrame(results).round(4)
display(oos_table)

# --- Plot cumulative trajectories vs the market benchmark ---
cum_mkt = (1 + ret_oos_mkt).cumprod()
fig, ax = plt.subplots()
for name, cum in cums.items():
    cum.plot(ax=ax, label=name)
cum_mkt.plot(ax=ax, label="SPY (market)", color="black", linestyle="--")
ax.set_title("Out-of-sample cumulative return (Jan-Mar 2026)")
ax.set_ylabel("Growth of $1"); ax.legend()
plt.tight_layout(); plt.show()
````

### Interpretations

1. Over the 13-week out-of-sample quarter, the market (SPY) fell by 4.4%, yet every portfolio delivered positive returns. The equal-weight portfolio returned 6.0%, the minimum-variance portfolio 9.6%, the tangency portfolio 10.9% and the bounded-short tangency portfolio 13.4%. The optimized portfolios therefore outperformed both the equal-weight benchmark and the market.

2. The minimum-variance portfolio behaved exactly as intended. It delivered the lowest realized volatility (11.3% compared to 18.8% for equal-weight), the smallest drawdown (-2.8%) and the highest Sharpe ratio (approximately 3.3). This provides strong evidence that the low-variance estimate was stable out of sample. The tangency portfolio was also reasonably stable, maintaining a positive Jensen's alpha with beta easing to approximately 1.4. The bounded-short version earned the most but with the highest volatility and drawdown, demonstrating that leverage cuts both ways.

3. In simple terms, the optimized portfolios worked well in this test period, but we should be cautious about reading too much into 13 weeks of data. The market happened to fall while defensive stocks held up, which played directly into the strengths of our optimized portfolios. This does not mean the strategy will always work. If the market had rallied strongly, the low-beta optimized portfolios might have lagged behind. The key takeaway is that the minimum-variance portfolio delivered on its promise of lower risk, while the tangency portfolio's success depended partly on favourable market conditions. Longer out-of-sample periods would be needed to draw firmer conclusions about whether these optimization techniques reliably add value.

## 8. Final Discussion

This analysis explored portfolio construction, diversification, mean-variance optimization and the CAPM using ten stocks selected from distinct GICS sectors. The most important lesson is that diversification benefits come from correlation, not from the number of holdings. The equal-weight portfolio achieved 20.1% volatility despite containing stocks with individual volatilities ranging from 21.8% to 53.6%, while the concentrated FCX and LRCX portfolio delivered 39.7% volatility because it paired two correlated cyclicals. Spreading capital across weakly correlated sectors is what drives risk reduction. The maximum drawdown analysis reinforced this point, with the concentrated portfolio suffering a 47% peak-to-trough loss compared to 28% for the diversified portfolio.

The sector-diversified stock selection proved valuable. Information Technology (LRCX) and Materials (FCX) delivered high returns but also high volatility and beta, behaving as expected for cyclical sectors. Consumer Staples (COST) and Utilities (NEE) provided stability with beta below 0.75, serving as portfolio anchors. Health Care (VRTX) and Communication Services (TTWO) exhibited idiosyncratic behaviour with low correlations, offering diversification benefits. Energy (EOG) and Consumer Discretionary (LULU) disappointed with high risk but inadequate returns, explaining their exclusion from optimized portfolios. Industrials (DE) and Financials (SPGI) occupied the middle ground with moderate risk-return profiles.

The mean-variance optimization produced economically sensible portfolios. The minimum-variance portfolio concentrated in COST, NEE, TTWO and DE, achieving 15.5% volatility and a Sharpe ratio of 1.34. The tangency portfolio accepted slightly more risk by adding LRCX exposure, achieving a Sharpe of 1.45. Both excluded high-volatility names whose risk was uncompensated. Bounded short-selling provided only marginal improvement, suggesting the long-only constraint was not restrictive for this universe. This outcome aligns with the theoretical expectation that short-selling helps most when some assets have negative expected returns or extremely high correlations, neither of which applied strongly here.

Out-of-sample results were encouraging but inconclusive. During Q1 2026, the minimum-variance portfolio returned 9.6% and the tangency 10.9% while SPY fell 4.4%. The minimum-variance portfolio delivered on its promise with lowest volatility and smallest drawdown. However, 13 weeks is too short for firm conclusions, and the defensive positioning happened to benefit from a falling market. In a strong bull market, these low-beta portfolios would likely underperform, highlighting the tradeoff between downside protection and upside participation.

Evidence for the CAPM was weak. High-beta stocks FCX and LRCX earned top returns, but the relationship was not monotonic as LULU and EOG underperformed lower-beta names. Most stocks plotted above the Security Market Line, showing positive alphas that likely represent in-sample artifacts rather than persistent anomalies.

Several limitations apply: the risk-free rate was assumed zero, the out-of-sample window was brief, historical estimates may not persist, transaction costs were ignored, and SPY captures only US large-caps. Despite these limitations, the analysis demonstrates that thoughtful sector diversification reduces risk, mean-variance optimization produces defensible portfolios, and out-of-sample testing remains essential before trusting any backtest.
